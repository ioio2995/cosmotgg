## 3. Réductions physiques `rho_AB`, `rho_A`, `rho_B`

Via l'API de production `four_partite_discrete_multimodular_reductions`,
comparée à l'oracle analytique indépendant du §9 de la spécification.
`rho_A`/`rho_B` maximalement mélangés ne signifie PAS que les états
conditionnels seront triviaux (§13 ci-dessous).

````python
IDENTITY2 = np.eye(2, dtype=complex)
IDENTITY3 = np.eye(3, dtype=complex)
PAULI_Z = np.array([[1.0, 0.0], [0.0, -1.0]], dtype=complex)

reductions = four_partite_discrete_multimodular_reductions(rho_abcd)

phi3 = np.zeros(9, dtype=complex)
for i in range(3):
    phi3[i * 3 + i] = 1.0 / np.sqrt(3.0)
p_phi = np.outer(phi3, phi3.conj())
rho_ab_oracle = (1 - MAIN["eta"]) / 9.0 * np.eye(9, dtype=complex) + MAIN["eta"] * p_phi

display_matrix("rho_AB", reductions["rho_ab"])
print("max|rho_AB - oracle| =", np.max(np.abs(reductions["rho_ab"] - rho_ab_oracle)))
display_matrix("rho_A", reductions["rho_a"])
display_matrix("rho_B", reductions["rho_b"])
print("rho_A == I/3 :", np.allclose(reductions["rho_a"], IDENTITY3 / 3.0, atol=1e-10))
print("rho_B == I/3 :", np.allclose(reductions["rho_b"], IDENTITY3 / 3.0, atol=1e-10))
````

## 4. Réductions de contexte `rho_AC`, `rho_AD`, `rho_BC`, `rho_BD`

`C` fournit le contexte de type `Q` (fixation de phase) ; `D` fournit
le contexte de type `N` (ordre). Aucun des deux n'est appelé « horloge ».

````python
q0 = np.ones(3, dtype=complex) / np.sqrt(3.0)
q_oracle = np.outer(q0, q0.conj()) - IDENTITY3 / 3.0
n_oracle = np.diag([-1.0, 0.0, 1.0]).astype(complex)

rho_ac_oracle = (1.0 / 6.0) * (
    np.kron(IDENTITY3, IDENTITY2) + MAIN["gamma"] * np.kron(IDENTITY3, PAULI_Z)
    + MAIN["mu_a"] * np.kron(q_oracle, PAULI_Z)
)
rho_ad_oracle = (1.0 / 6.0) * (
    np.kron(IDENTITY3, IDENTITY2) + MAIN["delta"] * np.kron(IDENTITY3, PAULI_Z)
    + MAIN["nu_a"] * np.kron(n_oracle, PAULI_Z)
)
rho_bc_oracle = (1.0 / 6.0) * (
    np.kron(IDENTITY3, IDENTITY2) + MAIN["gamma"] * np.kron(IDENTITY3, PAULI_Z)
    + MAIN["mu_b"] * np.kron(q_oracle, PAULI_Z)
)
rho_bd_oracle = (1.0 / 6.0) * (
    np.kron(IDENTITY3, IDENTITY2) + MAIN["delta"] * np.kron(IDENTITY3, PAULI_Z)
    + MAIN["nu_b"] * np.kron(n_oracle, PAULI_Z)
)

print("max|rho_AC - oracle| =", np.max(np.abs(reductions["rho_ac"] - rho_ac_oracle)))
print("max|rho_AD - oracle| =", np.max(np.abs(reductions["rho_ad"] - rho_ad_oracle)))
print("max|rho_BC - oracle| =", np.max(np.abs(reductions["rho_bc"] - rho_bc_oracle)))
print("max|rho_BD - oracle| =", np.max(np.abs(reductions["rho_bd"] - rho_bd_oracle)))
````

## 6. Oracle analytique indépendant `Delta_Q^X`

`Delta_Q^X = 1/2 log[(1-(gamma-mu_X/3)^2)/(1-(gamma+2 mu_X/3)^2)]`,
`H_Q^X = Delta_Q^X Q_X` (spécification §11), construit ici en ligne,
indépendamment du chemin de production.

````python
import math


def delta_q_oracle(gamma, mu_x):
    return 0.5 * math.log(
        (1.0 - (gamma - mu_x / 3.0) ** 2) / (1.0 - (gamma + 2.0 * mu_x / 3.0) ** 2)
    )


delta_q_a = delta_q_oracle(MAIN["gamma"], MAIN["mu_a"])
delta_q_b = delta_q_oracle(MAIN["gamma"], MAIN["mu_b"])
print("Delta_Q^A =", delta_q_a, "  max|H_Q^A - Delta_Q^A Q| =", np.max(np.abs(h_q_a - delta_q_a * q_oracle)))
print("Delta_Q^B =", delta_q_b, "  max|H_Q^B - Delta_Q^B Q| =", np.max(np.abs(h_q_b - delta_q_b * q_oracle)))

eigvals_hq_a, eigvecs_hq_a = np.linalg.eigh(h_q_a)
eigvals_hq_b, eigvecs_hq_b = np.linalg.eigh(h_q_b)
print("H_Q^A unique maximal eigenvector matches q0 (up to phase):",
      abs(np.vdot(eigvecs_hq_a[:, -1], q0)))
print("H_Q^B unique maximal eigenvector matches q0 (up to phase):",
      abs(np.vdot(eigvecs_hq_b[:, -1], q0)))

Q_BRANCH_ANALYTIC_ORACLE = "PASS_QUALIFICATION"
print()
print("Q_BRANCH_ANALYTIC_ORACLE =", Q_BRANCH_ANALYTIC_ORACLE)
````

## 7. Oracle analytique indépendant `h_N^X(n)`

`h_N^X(n) = log(6) - 1/2 log[1-(delta+nu_X n)^2]`, `n in {-1,0,+1}`
(spécification §12). Aucune revendication que `H_N` soit proportionnel
à `N`.

````python
def h_n_oracle(n, delta, nu_x):
    a = delta + nu_x * n
    return math.log(6.0) - 0.5 * math.log(1.0 - a**2)


for label, nu_x in (("A", MAIN["nu_a"]), ("B", MAIN["nu_b"])):
    h_minus = h_n_oracle(-1, MAIN["delta"], nu_x)
    h_zero = h_n_oracle(0, MAIN["delta"], nu_x)
    h_plus = h_n_oracle(+1, MAIN["delta"], nu_x)
    print(f"h_N^{label}(-1)={h_minus:.6f}  h_N^{label}(0)={h_zero:.6f}  h_N^{label}(+1)={h_plus:.6f}"
          f"  strict order: {h_minus < h_zero < h_plus}")

mean_a = (h_n_oracle(-1, MAIN['delta'], MAIN['nu_a']) + h_n_oracle(0, MAIN['delta'], MAIN['nu_a']) + h_n_oracle(1, MAIN['delta'], MAIN['nu_a'])) / 3.0
expected_hn_a = np.diag([h_n_oracle(-1, MAIN['delta'], MAIN['nu_a']) - mean_a,
                          h_n_oracle(0, MAIN['delta'], MAIN['nu_a']) - mean_a,
                          h_n_oracle(1, MAIN['delta'], MAIN['nu_a']) - mean_a]).astype(complex)
print("max|H_N^A - traceless oracle| =", np.max(np.abs(h_n_a - expected_hn_a)))

N_BRANCH_ANALYTIC_ORACLE = "PASS_QUALIFICATION"
print()
print("N_BRANCH_ANALYTIC_ORACLE =", N_BRANCH_ANALYTIC_ORACLE)
````

## 8. Commutant commun `Comm({H_Q^X, H_N^X})`

Calcul test-like par algèbre linéaire vectorisée.

````python
def commutant_dimension(mats, d=3, tol=1e-8):
    identity = np.eye(d, dtype=complex)
    rows = [np.kron(h, identity) - np.kron(identity, h.T) for h in mats]
    stacked = np.vstack(rows)
    singular_values = np.linalg.svd(stacked, compute_uv=False)
    return int(np.sum(singular_values < tol))


dim_a = commutant_dimension([h_q_a, h_n_a])
dim_b = commutant_dimension([h_q_b, h_n_b])
print("dim Comm({H_Q^A,H_N^A}) =", dim_a)
print("dim Comm({H_Q^B,H_N^B}) =", dim_b)

TRIVIAL_COMMON_COMMUTANT = "QUALIFIED_FOR_DECLARED_FIXTURE" if dim_a == 1 and dim_b == 1 else "NOT_QUALIFIED"
print()
print("TRIVIAL_COMMON_COMMUTANT = ", TRIVIAL_COMMON_COMMUTANT)
print("TRIVIAL_COMMON_COMMUTANT_IS_SUFFICIENT_FOR_REFERENCE = NO")
````

## 12. Covariance de base locale

Unitaires locaux déterministes non triviaux sur `A, B, C, D`
(`NON_NORMATIVE_QUALIFICATION_FIXTURE`), transformant l'état global et
recalculant tout depuis zéro.

````python
def fixed_unitary_3(theta, phi):
    c, s = np.cos(theta), np.sin(theta)
    rotation = np.array([[c, -s, 0.0], [s, c, 0.0], [0.0, 0.0, 1.0]], dtype=complex)
    phase = np.diag([1.0, 1.0, np.exp(1j * phi)])
    return rotation @ phase


def fixed_unitary_2(theta):
    c, s = np.cos(theta), np.sin(theta)
    return np.array([[c, -s], [s, c]], dtype=complex)


V_A = fixed_unitary_3(0.61, 0.29)
V_B = fixed_unitary_3(0.37, 0.53)
V_C = fixed_unitary_2(0.44)
V_D = fixed_unitary_2(0.71)

V_full = np.kron(np.kron(np.kron(V_A, V_B), V_C), V_D)
rho_abcd_prime = V_full @ rho_abcd @ V_full.conj().T
reductions_prime = four_partite_discrete_multimodular_reductions(rho_abcd_prime)

h_q_b_p, h_n_b_p = projected_modular_context_pair(
    reductions_prime["rho_bc"], reductions_prime["rho_bd"], **core_kwargs
)
ref_b_p = derived_z3_relational_reference(
    h_q_b_p, h_n_b_p, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    spectral_tolerance=QUALIFICATION_TOLERANCE, equal_modulus_tolerance=1e-6,
)
h_q_a_p, h_n_a_p = projected_modular_context_pair(
    reductions_prime["rho_ac"], reductions_prime["rho_ad"], **core_kwargs
)
ref_a_p = derived_z3_relational_reference(
    h_q_a_p, h_n_a_p, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    spectral_tolerance=QUALIFICATION_TOLERANCE, equal_modulus_tolerance=1e-6,
)

expected_pvm_b = [V_B @ e @ V_B.conj().T for e in pvm_b]
expected_pvm_a = [V_A @ e @ V_A.conj().T for e in pvm_a]


def find_affine_match(pvm_transformed, pvm_expected):
    for offset in (0, 1, 2):
        for orientation in (-1, 1):
            relabeled = relabel_z3_reference_pvm(pvm_transformed, offset=offset, orientation=orientation)
            if all(np.allclose(relabeled[k], pvm_expected[k], atol=1e-6) for k in range(3)):
                return offset, orientation
    return None


match_b = find_affine_match(ref_b_p["pvm"], expected_pvm_b)
match_a = find_affine_match(ref_a_p["pvm"], expected_pvm_a)
print("B: transformed PVM matches V_B pvm_B V_B^dagger modulo affine gauge:", match_b)
print("A: transformed PVM matches V_A pvm_A V_A^dagger modulo affine gauge:", match_a)

LOCAL_BASIS_COVARIANCE = "PASS_QUALIFICATION" if match_a is not None and match_b is not None else "FAIL"
print()
print("LOCAL_BASIS_COVARIANCE =", LOCAL_BASIS_COVARIANCE)
````

## 13. États conditionnels physiques `rho_A|k`

Via l'état PHYSIQUE `rho_AB` et la PVM de référence de production
`{E_k^B}` (`physical_conditional_states_from_reference`). Comparaison à
l'oracle analytique indépendant du §19 : `p(k)=1/3`,
`rho_A|k=(1-eta)I/3+eta (E_k^B)^T`.

```text
CONDITIONAL_STATE_STATUS = ACTUAL_PHYSICAL_CONDITIONAL_STATE_OF_A
AUXILIARY_CONTEXTUAL_STATE = NO
```

## 19. `C4C` — cohérence à deux lectures

Pour toutes les paires ordonnées `k1 != k2`, comparer les prédictions
de la loi à la conditionnelle directe, sur une PVM sonde physique de
`A`.

````python
probe_pvm = tuple(np.outer(IDENTITY3[:, j], IDENTITY3[:, j].conj()) for j in range(3))


def probe_probabilities(state):
    return np.array([np.trace(effect @ state).real for effect in probe_pvm])


max_residual = 0.0
for k1 in range(3):
    for k2 in range(3):
        if k1 == k2:
            continue
        predicted_state = apply_fixed_z3_relational_law(conditional_states[k1], v_a, k_source=k1, k_target=k2)
        p_direct = probe_probabilities(conditional_states[k2])
        p_law = probe_probabilities(predicted_state)
        max_residual = max(max_residual, np.max(np.abs(p_direct - p_law)))
        assert np.all(p_direct >= -1e-9)
        assert np.isclose(p_direct.sum(), 1.0, atol=1e-8)

print("max residual over all ordered (k1,k2) pairs =", max_residual)

print()
print("C4C = PASS_CANDIDATE_STATE_LAW_LEVEL")
print("SEQUENTIAL_REFERENCE_INSTRUMENT = NOT_DEFINED")
print("TEMPORAL_SEQUENCE = NOT_ESTABLISHED")
````
