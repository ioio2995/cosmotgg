## 3. Construction SU(2) gelée (plan §V2) et résidu normalisé (plan §V9)

$$
U(\alpha;\vec n) = \cos(\alpha/2)\,I - i\sin(\alpha/2)\,
\frac{n_x\sigma_x+n_y\sigma_y+n_z\sigma_z}{\|\vec n\|}.
$$

$$
R(A,B) = \frac{\|A-B\|_F}{\max(1,\ \|A\|_F,\ \|B\|_F)}.
$$

Construction unitaire déterministe uniquement ; aucune interprétation
d'angle physique ni d'axe spatial.

Câblage déterministe des tolérances vers les appels de production
(plan §V8) : `fine_relational_hamiltonian(max_entanglement_unitarity_tolerance=
UNITARY_INPUT_TOLERANCE)`, `fine_relational_gibbs_state`/`modular_datum`
`(hermiticity_tolerance=HERMITICITY_TOLERANCE, trace_tolerance=
TRACE_TOLERANCE, positivity_tolerance=POSITIVITY_TOLERANCE)`.


````python
from cosmotgg.core.states import embed_operator
from cosmotgg.models.model1b import directional, hierarchy, modular_support, states

PAULI = modular_support.PAULI_STACK  # {I, X, Y, Z}
FINE_EDGES = states.FINE_EDGES
FINE_SITE_ORDER = hierarchy.FINE_SITE_ORDER
LEVEL_1_SITES = hierarchy.LEVEL_1_SITES
LEVEL_0_SITES = hierarchy.LEVEL_0_SITES


def U(alpha, n):
    """U(alpha; n) = cos(alpha/2) I - i sin(alpha/2) (n . sigma)/|n| (plan V2)."""
    n_arr = np.array(n, dtype=float)
    n_hat = n_arr / np.linalg.norm(n_arr)
    identity2 = np.eye(2, dtype=complex)
    sigma_n = n_hat[0] * PAULI[1] + n_hat[1] * PAULI[2] + n_hat[2] * PAULI[3]
    return np.cos(alpha / 2.0) * identity2 - 1j * np.sin(alpha / 2.0) * sigma_n


def R(a, b):
    """R(A,B) = ||A-B||_F / max(1, ||A||_F, ||B||_F) (plan V9)."""
    a = np.asarray(a)
    b = np.asarray(b)
    return float(np.linalg.norm(a - b) / max(1.0, np.linalg.norm(a), np.linalg.norm(b)))


def active_cycle_edges(level_sites):
    """Cycle actif Gamma_n: (v_0,v_1),(v_1,v_2),...,(v_{m-1},v_0), v_0=A."""
    n = len(level_sites)
    return [(level_sites[k], level_sites[(k + 1) % n]) for k in range(n)]


def compute_j_o_q(k_matrix, n_sites, level_sites):
    """Blocs J/O actifs (route J -> directional_factor(J)) et Q_n via la
    route de production réelle active_cycle_loop_object_from_blocks."""
    pos = {label: i for i, label in enumerate(level_sites)}
    edges = active_cycle_edges(level_sites)
    j_blocks, o_blocks = {}, {}
    for (vi, vj) in edges:
        j = modular_support.global_two_body_block(k_matrix, n_sites, pos[vi], pos[vj])
        j_blocks[(vi, vj)] = j
        o_blocks[(vi, vj)] = directional.directional_factor(j)
    ordered_j = [j_blocks[e] for e in edges]
    q = directional.active_cycle_loop_object_from_blocks(ordered_j)
    return j_blocks, o_blocks, q, edges


ALL_O_BLOCKS = []   # (label, matrix) -- orthogonality guard (section 19)
ALL_Q_BLOCKS = []   # (label, matrix) -- orthogonality guard (section 19)
ALL_J_REPORTS = []  # (label, matrix) -- conditioning report (section 20)
RAW_RESIDUALS = []  # (label, value, tolerance_name) -- residual table (section 21)
VERDICTS = {}       # criterion -> (status, detail)

print("U, R, active_cycle_edges, compute_j_o_q définis.")
print("FINE_EDGES      =", FINE_EDGES)
print("FINE_SITE_ORDER =", FINE_SITE_ORDER)
print("LEVEL_1_SITES   =", LEVEL_1_SITES)
print("LEVEL_0_SITES   =", LEVEL_0_SITES)
````

## 4. Fixture générique non centrale V3 (plan §V3)

$$
\theta_i = 0.10+0.02i,\quad \alpha_i = 0.30+0.10i,\quad
\vec n_i = (1,\,1+(i\bmod3),\,2),\quad M_{e_i}=U(\alpha_i;\vec n_i),
$$

pour l'ordre d'arête fixe `AX, XY, YB, BC, CP, PQ, QD, DA`.

Contrôle de non-centralité préenregistré :
$H_M = M_{AX}\overline{M_{XY}}M_{YB}\overline{M_{BC}}M_{CP}\overline{M_{PQ}}M_{QD}\overline{M_{DA}}$,
$r_{\mathrm{noncentral}} = \|H_M - \tfrac{\mathrm{Tr}(H_M)}2 I_2\|_F > \mathrm{SIGNAL\_FLOOR}$.
Si cet échec survient, la fixture n'est PAS remplacée
(`GENERIC_FIXTURE = PREDECLARED_FIXTURE_NOT_NONCENTRAL`).


````python
thetas_v3 = {edge: 0.10 + 0.02 * i for i, edge in enumerate(FINE_EDGES)}
ns_v3 = {edge: (1, 1 + (i % 3), 2) for i, edge in enumerate(FINE_EDGES)}
alphas_v3 = {edge: 0.30 + 0.10 * i for i, edge in enumerate(FINE_EDGES)}
Ms_v3 = {edge: U(alphas_v3[edge], ns_v3[edge]) for edge in FINE_EDGES}

print("theta_i (V3):", thetas_v3)

H_M = (
    Ms_v3["AX"] @ Ms_v3["XY"].conj() @ Ms_v3["YB"] @ Ms_v3["BC"].conj()
    @ Ms_v3["CP"] @ Ms_v3["PQ"].conj() @ Ms_v3["QD"] @ Ms_v3["DA"].conj()
)
r_noncentral = float(np.linalg.norm(H_M - (np.trace(H_M) / 2.0) * np.eye(2, dtype=complex)))
print()
print("r_noncentral (RAW) =", r_noncentral)
print("SIGNAL_FLOOR        =", SIGNAL_FLOOR)

if r_noncentral > SIGNAL_FLOOR:
    GENERIC_FIXTURE = "ADMISSIBLE_NONCENTRAL"
else:
    GENERIC_FIXTURE = "PREDECLARED_FIXTURE_NOT_NONCENTRAL"
print("GENERIC_FIXTURE =", GENERIC_FIXTURE)
RAW_RESIDUALS.append(("r_noncentral (V3)", r_noncentral, "SIGNAL_FLOOR"))
````

## 5. Construction `rho2`/`rho1`/`rho0` (V3), `K_n`, blocs `J`/`O` actifs, `Q_n`, `chi_n`

`K1_v3`, `K0_v3`, `K0_direct_v3`, `Q2_v3`, `Q1_v3`, `Q0_v3`) sont réutilisés
par `T5F3`–`T5F11` ci-dessous.


````python
h_rel_v3 = states.fine_relational_hamiltonian(
    thetas_v3, Ms_v3, max_entanglement_unitarity_tolerance=UNITARY_INPUT_TOLERANCE
)
rho2_v3 = states.fine_relational_gibbs_state(
    h_rel_v3,
    hermiticity_tolerance=HERMITICITY_TOLERANCE,
    trace_tolerance=TRACE_TOLERANCE,
    positivity_tolerance=POSITIVITY_TOLERANCE,
)
rho1_v3 = hierarchy.reduce_to_level_1(rho2_v3)
rho0_v3 = hierarchy.reduce_to_level_0(rho1_v3)
rho0_direct_v3 = hierarchy.reduce_to_level_0_direct(rho2_v3)

print("rho2_v3 shape/trace:", rho2_v3.shape, np.trace(rho2_v3))
print("rho1_v3 shape/trace:", rho1_v3.shape, np.trace(rho1_v3))
print("rho0_v3 shape/trace:", rho0_v3.shape, np.trace(rho0_v3))
print("rho0_direct_v3 shape/trace:", rho0_direct_v3.shape, np.trace(rho0_direct_v3))

_modular_kwargs = dict(
    hermiticity_tolerance=HERMITICITY_TOLERANCE,
    trace_tolerance=TRACE_TOLERANCE,
    positivity_tolerance=POSITIVITY_TOLERANCE,
)
K2_v3 = modular_support.modular_datum(rho2_v3, **_modular_kwargs)
K1_v3 = modular_support.modular_datum(rho1_v3, **_modular_kwargs)
K0_v3 = modular_support.modular_datum(rho0_v3, **_modular_kwargs)
K0_direct_v3 = modular_support.modular_datum(rho0_direct_v3, **_modular_kwargs)
print()
print("K2_v3/K1_v3/K0_v3/K0_direct_v3 shapes:", K2_v3.shape, K1_v3.shape, K0_v3.shape, K0_direct_v3.shape)

j2_v3, o2_v3, Q2_v3, edges2 = compute_j_o_q(K2_v3, 8, FINE_SITE_ORDER)
j1_v3, o1_v3, Q1_v3, edges1 = compute_j_o_q(K1_v3, 6, LEVEL_1_SITES)
j0_v3, o0_v3, Q0_v3, edges0 = compute_j_o_q(K0_v3, 4, LEVEL_0_SITES)

for level_name, edges, jb, ob in [("level2", edges2, j2_v3, o2_v3), ("level1", edges1, j1_v3, o1_v3), ("level0", edges0, j0_v3, o0_v3)]:
    for e in edges:
        ALL_J_REPORTS.append((f"V3 {level_name} J_{e}", jb[e]))
        ALL_O_BLOCKS.append((f"V3 {level_name} O_{e}", ob[e]))
ALL_Q_BLOCKS.append(("V3 Q2", Q2_v3))
ALL_Q_BLOCKS.append(("V3 Q1", Q1_v3))
ALL_Q_BLOCKS.append(("V3 Q0", Q0_v3))

chi2_v3 = directional.conjugacy_class_scalar(Q2_v3)
chi1_v3 = directional.conjugacy_class_scalar(Q1_v3)
chi0_v3 = directional.conjugacy_class_scalar(Q0_v3)
print()
print("Q2_v3 =\n", Q2_v3)
print("Q1_v3 =\n", Q1_v3)
print("Q0_v3 =\n", Q0_v3)
print("chi2_v3, chi1_v3, chi0_v3 =", chi2_v3, chi1_v3, chi0_v3)
````

## 10. `T5F5` — Support modulaire complet (plan §V13)

## 18. Oracle négatif d'arbre `8->6` / `6->4` (plan §V5, §V20, fixture V5)

Mêmes arêtes/valeurs que V3 sauf `theta_DA=0`, `M_DA=I_2`. Chemins :
`O_path,8->6 = O_CP O_PQ O_QD` comparé à `O_CD^(1)` ; `O_path,6->4 =
O_AX^(1) O_XY^(1) O_YB^(1)` comparé à `O_AB^(0)`.
$\mathrm{tree\_residual} = \|D_{\mathrm{tree}}-I\|_F/\sqrt8 \le
\mathrm{TREE\_AGREEMENT\_TOLERANCE}$.


````python
thetas_v5 = dict(thetas_v3)
thetas_v5["DA"] = 0.0
Ms_v5 = dict(Ms_v3)
Ms_v5["DA"] = np.eye(2, dtype=complex)

h_rel_v5 = states.fine_relational_hamiltonian(
    thetas_v5, Ms_v5, max_entanglement_unitarity_tolerance=UNITARY_INPUT_TOLERANCE
)
rho2_v5 = states.fine_relational_gibbs_state(h_rel_v5, **_modular_kwargs)
rho1_v5 = hierarchy.reduce_to_level_1(rho2_v5)
rho0_v5 = hierarchy.reduce_to_level_0(rho1_v5)
K2_v5 = modular_support.modular_datum(rho2_v5, **_modular_kwargs)
K1_v5 = modular_support.modular_datum(rho1_v5, **_modular_kwargs)
K0_v5 = modular_support.modular_datum(rho0_v5, **_modular_kwargs)

J_CP = modular_support.global_two_body_block(K2_v5, 8, pos2["C"], pos2["P"])
J_PQ = modular_support.global_two_body_block(K2_v5, 8, pos2["P"], pos2["Q"])
J_QD = modular_support.global_two_body_block(K2_v5, 8, pos2["Q"], pos2["D"])
O_CP = directional.directional_factor(J_CP)
O_PQ = directional.directional_factor(J_PQ)
O_QD = directional.directional_factor(J_QD)
O_path_8to6 = O_CP @ O_PQ @ O_QD

J_CD_1 = modular_support.global_two_body_block(K1_v5, 6, pos1["C"], pos1["D"])
O_CD_1 = directional.directional_factor(J_CD_1)

D_tree_8to6 = O_path_8to6.T @ O_CD_1
tree_residual_8to6 = float(np.linalg.norm(D_tree_8to6 - np.eye(3)) / np.sqrt(8.0))
print("D_tree_8to6 [RAW] =\n", D_tree_8to6)
print("tree_residual_8to6 [RAW] =", tree_residual_8to6)
RAW_RESIDUALS.append(("tree_residual_8to6", tree_residual_8to6, "TREE_AGREEMENT_TOLERANCE"))

J_AX_1 = modular_support.global_two_body_block(K1_v5, 6, pos1["A"], pos1["X"])
J_XY_1 = modular_support.global_two_body_block(K1_v5, 6, pos1["X"], pos1["Y"])
J_YB_1 = modular_support.global_two_body_block(K1_v5, 6, pos1["Y"], pos1["B"])
O_AX_1 = directional.directional_factor(J_AX_1)
O_XY_1 = directional.directional_factor(J_XY_1)
O_YB_1 = directional.directional_factor(J_YB_1)
O_path_6to4 = O_AX_1 @ O_XY_1 @ O_YB_1

J_AB_0 = modular_support.global_two_body_block(K0_v5, 4, pos0["A"], pos0["B"])
O_AB_0 = directional.directional_factor(J_AB_0)

D_tree_6to4 = O_path_6to4.T @ O_AB_0
tree_residual_6to4 = float(np.linalg.norm(D_tree_6to4 - np.eye(3)) / np.sqrt(8.0))
print("D_tree_6to4 [RAW] =\n", D_tree_6to4)
print("tree_residual_6to4 [RAW] =", tree_residual_6to4)
RAW_RESIDUALS.append(("tree_residual_6to4", tree_residual_6to4, "TREE_AGREEMENT_TOLERANCE"))

for lbl, mat in [("V5 O_CP", O_CP), ("V5 O_PQ", O_PQ), ("V5 O_QD", O_QD), ("V5 O_CD_1", O_CD_1),
                  ("V5 O_AX_1", O_AX_1), ("V5 O_XY_1", O_XY_1), ("V5 O_YB_1", O_YB_1), ("V5 O_AB_0", O_AB_0)]:
    ALL_O_BLOCKS.append((lbl, mat))
for lbl, mat in [("V5 J_CP", J_CP), ("V5 J_PQ", J_PQ), ("V5 J_QD", J_QD), ("V5 J_CD_1", J_CD_1),
                  ("V5 J_AX_1", J_AX_1), ("V5 J_XY_1", J_XY_1), ("V5 J_YB_1", J_YB_1), ("V5 J_AB_0", J_AB_0)]:
    ALL_J_REPORTS.append((lbl, mat))

tree_ok = (tree_residual_8to6 <= TREE_AGREEMENT_TOLERANCE) and (tree_residual_6to4 <= TREE_AGREEMENT_TOLERANCE)
TREE_DIRECTIONAL_RUNNING = "ABSENT" if tree_ok else "PRESENT"
print()
print("TREE_DIRECTIONAL_RUNNING =", TREE_DIRECTIONAL_RUNNING)
````

## 19. Contrôles mécaniques d'orthogonalité (plan §V21)

Gardes d'implémentation uniquement. Pour tout `O` : `||O^T O - I||_F \le
ORTHOGONALITY_REGRESSION_TOLERANCE`. Pour tout `Q` : idem, et
`det(Q) > 0`.


````python
orth_o_residuals = []
for label, o_matrix in ALL_O_BLOCKS:
    res = float(np.linalg.norm(o_matrix.T @ o_matrix - np.eye(3)))
    orth_o_residuals.append((label, res))
max_o_residual = max(v for _, v in orth_o_residuals)
print(f"{len(orth_o_residuals)} facteurs O contrôlés ; max||O^T O - I||_F [RAW] = {max_o_residual:.3e}")

orth_q_residuals = []
det_q_signs = []
for label, q_matrix in ALL_Q_BLOCKS:
    res = float(np.linalg.norm(q_matrix.T @ q_matrix - np.eye(3)))
    det_q = float(np.linalg.det(q_matrix))
    orth_q_residuals.append((label, res))
    det_q_signs.append((label, det_q))
max_q_residual = max(v for _, v in orth_q_residuals)
print(f"{len(orth_q_residuals)} objets Q contrôlés ; max||Q^T Q - I||_F [RAW] = {max_q_residual:.3e}")
print("det(Q) pour chaque Q [RAW] =", det_q_signs)

orthogonality_ok = (
    max_o_residual <= ORTHOGONALITY_REGRESSION_TOLERANCE
    and max_q_residual <= ORTHOGONALITY_REGRESSION_TOLERANCE
    and all(d > 0 for _, d in det_q_signs)
)
print("Contrôles mécaniques d'orthogonalité : PASS =" , orthogonality_ok)
````
