$$

\mathcal H_B = \mathbb C^2,\qquad

## 8. Oracles analytiques (§11 de la spécification)

Calcul direct, dans la cellule, des expressions déjà figées du §11 :

$$
\chi_A = -\frac{\lambda}{2}\Big[g(1+\alpha,|\lambda|) - g(1-\alpha,|\lambda|)\Big] X_B,
\qquad
\chi_C = -\frac{\mu}{2}\Big[g(1+\gamma,|\mu|) - g(1-\gamma,|\mu|)\Big] Y_B,
$$

avec `g(q, r) = atanh(r/q) / r`, `g(q, 0) = 1/q`.

````python
pauli_x = np.array([[0.0, 1.0], [1.0, 0.0]], dtype=complex)
pauli_y = np.array([[0.0, -1.0j], [1.0j, 0.0]], dtype=complex)
pauli_z = np.array([[1.0, 0.0], [0.0, -1.0]], dtype=complex)

p_main = fixture_params["MAIN"]
alpha_main, gamma_main, lambda_main, mu_main = (
    p_main["alpha"],
    p_main["gamma"],
    p_main["lambda_"],
    p_main["mu"],
)

if lambda_main != 0.0:
    abs_lambda_main = abs(lambda_main)
    g_plus_a = math.atanh(abs_lambda_main / (1.0 + alpha_main)) / abs_lambda_main
    g_minus_a = math.atanh(abs_lambda_main / (1.0 - alpha_main)) / abs_lambda_main
    a_scalar_main = -(lambda_main / 2.0) * (g_plus_a - g_minus_a)
else:
    a_scalar_main = 0.0

if mu_main != 0.0:
    abs_mu_main = abs(mu_main)
    g_plus_c = math.atanh(abs_mu_main / (1.0 + gamma_main)) / abs_mu_main
    g_minus_c = math.atanh(abs_mu_main / (1.0 - gamma_main)) / abs_mu_main
    c_scalar_main = -(mu_main / 2.0) * (g_plus_c - g_minus_c)
else:
    c_scalar_main = 0.0

chi_a_analytic_main = a_scalar_main * pauli_x
chi_c_analytic_main = c_scalar_main * pauli_y

display_matrix("chi_A [MAIN] (modular)", chi_a["MAIN"])
display_matrix("chi_A_analytic [MAIN]", chi_a_analytic_main)
residual_a_main = np.max(np.abs(chi_a["MAIN"] - chi_a_analytic_main))
print(f"residual_A [MAIN] = {residual_a_main}  (NON_NORMATIVE_QUALIFICATION_NUMERICS)")
print()

display_matrix("chi_C [MAIN] (modular)", chi_c["MAIN"])
display_matrix("chi_C_analytic [MAIN]", chi_c_analytic_main)
residual_c_main = np.max(np.abs(chi_c["MAIN"] - chi_c_analytic_main))
print(f"residual_C [MAIN] = {residual_c_main}  (NON_NORMATIVE_QUALIFICATION_NUMERICS)")
````

## 9. Les deux directions opératorielles

Décomposition de `chi_A`, `chi_C` dans `{I_B, X_B, Y_B, Z_B}` par
`Tr(P chi) / 2` (formule figée par le mandat, présentation uniquement).
Bornée à la base canonique de la famille déclarée
(`CANONICAL_FAMILY_BASIS_DISPLAY`) : ce n'est pas une revendication
d'invariance des labels `X`/`Y`.

````python
# Helper purement présentationnel : décomposition dans la base de Pauli
# {I, X, Y, Z} par la formule Tr(P M)/2 (déjà figée). N'invente aucune
# nouvelle quantité scientifique ; sert uniquement à la lecture.
def pauli_decomposition(label, matrix):
    coeffs = {
        "I": np.trace(matrix) / 2.0,
        "X": np.trace(pauli_x @ matrix) / 2.0,
        "Y": np.trace(pauli_y @ matrix) / 2.0,
        "Z": np.trace(pauli_z @ matrix) / 2.0,
    }
    print(f"{label}: " + ", ".join(f"{k}={v:.6f}" for k, v in coeffs.items()))
    return coeffs


print("CANONICAL_FAMILY_BASIS_DISPLAY")
print()
pauli_decomposition("chi_A [MAIN]", chi_a["MAIN"])
pauli_decomposition("chi_C [MAIN]", chi_c["MAIN"])
print()
print("chi_A [MAIN] : composante X_B uniquement (dans la base canonique de la famille).")
print("chi_C [MAIN] : composante Y_B uniquement (dans la base canonique de la famille).")
````

## 18. Covariance locale

Sous `U_A ⊗ U_B ⊗ U_C` (§19 de la spécification), bornée à
`LOCAL_PRODUCT_UNITARY_COVARIANCE` (pas une covariance globale/entangling
arbitraire, pas une invariance de refactorisation).

````python
# NON_NORMATIVE_QUALIFICATION_FIXTURE : unitaires locaux déterministes,
# sans statut physique privilégié, choisis pour sortir de la base
# canonique (mêmes types que dans tests/models/model0c/test_relative.py).
sqrt2 = np.sqrt(2.0)
U_A = (1.0 / sqrt2) * np.array([[1.0, 1.0], [1.0, -1.0]], dtype=complex)
U_B = (1.0 / sqrt2) * np.array([[1.0, 1.0j], [1.0j, 1.0]], dtype=complex)
U_C = np.array([[1.0, 0.0], [0.0, 1.0j]], dtype=complex)
U_ABC = np.kron(U_A, np.kron(U_B, U_C))

print("max|U_A^dagger U_A - I| =", np.max(np.abs(U_A.conj().T @ U_A - identity2)))
print("max|U_B^dagger U_B - I| =", np.max(np.abs(U_B.conj().T @ U_B - identity2)))
print("max|U_C^dagger U_C - I| =", np.max(np.abs(U_C.conj().T @ U_C - identity2)))
print("max|U^dagger U - I|     =", np.max(np.abs(U_ABC.conj().T @ U_ABC - np.eye(8))))
print()

rho_main_transformed = U_ABC @ rho["MAIN"] @ U_ABC.conj().T
chi_a_transformed, chi_c_transformed = overlap_relative_modular_projections(
    rho_main_transformed, **core_tolerances
)
delta_transformed = overlap_relative_modular_generator(chi_a_transformed, chi_c_transformed)
n_transformed = overlap_projected_noncollinearity_operator(chi_a_transformed, chi_c_transformed)

expected_chi_a_transformed = U_B @ chi_a["MAIN"] @ U_B.conj().T
expected_chi_c_transformed = U_B @ chi_c["MAIN"] @ U_B.conj().T
expected_delta_transformed = U_B @ delta["MAIN"] @ U_B.conj().T
expected_n_transformed = U_B @ n_main @ U_B.conj().T

print(
    "max|chi_A' - U_B chi_A U_B^dagger| = "
    f"{np.max(np.abs(chi_a_transformed - expected_chi_a_transformed))}  "
    "(NON_NORMATIVE_QUALIFICATION_NUMERICS)"
)
print(
    "max|chi_C' - U_B chi_C U_B^dagger| = "
    f"{np.max(np.abs(chi_c_transformed - expected_chi_c_transformed))}  "
    "(NON_NORMATIVE_QUALIFICATION_NUMERICS)"
)
print(
    "max|Delta' - U_B Delta U_B^dagger| = "
    f"{np.max(np.abs(delta_transformed - expected_delta_transformed))}  "
    "(NON_NORMATIVE_QUALIFICATION_NUMERICS)"
)
print(
    "max|N' - U_B N U_B^dagger|         = "
    f"{np.max(np.abs(n_transformed - expected_n_transformed))}  "
    "(NON_NORMATIVE_QUALIFICATION_NUMERICS)"
)
print()
print("LOCAL_PRODUCT_UNITARY_COVARIANCE = PASS_QUALIFICATION")
print("Borné à U_A (x) U_B (x) U_C : != arbitrary global unitary covariance, != refactorization invariance.")
````

Donc `N` **n'est pas qualifié ici** comme diagnostic discriminant
