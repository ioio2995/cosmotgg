Aucune interprétation physique de la magnitude de `epsilon`.

````python
from cosmotgg.models.model1a.states import (
    four_qubit_relational_loop_state,
    four_qubit_relational_loop_reductions,
)
from cosmotgg.models.model1a.links import (
    state_derived_edge_link,
    apply_directional_link,
    reverse_correlation_matrix,
    state_derived_centered_edge_transfer,
)
from cosmotgg.models.model1a.loop import (
    projective_loop_holonomy,
    projective_loop_action,
    state_derived_loop_transfer,
    relational_curvature_response_candidate,
)

IDENTITY2 = np.eye(2, dtype=complex)
IDENTITY4 = np.eye(4, dtype=complex)
IDENTITY16 = np.eye(16, dtype=complex)
PAULI_X = np.array([[0.0, 1.0], [1.0, 0.0]], dtype=complex)
PAULI_Y = np.array([[0.0, -1.0j], [1.0j, 0.0]], dtype=complex)
PAULI_Z = np.array([[1.0, 0.0], [0.0, -1.0]], dtype=complex)

# NON_NORMATIVE_QUALIFICATION_NUMERICS: tolerance used to call production
# primitives below. Not a physical threshold, not a T2/T4 threshold, not
# an acceptance criterion.
QUALIFICATION_TOLERANCE = 1e-9
UNITARITY_TOLERANCE = 1e-8

core_kwargs = dict(
    max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
    hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE,
    positivity_tolerance=QUALIFICATION_TOLERANCE,
)
link_kwargs = dict(
    hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE,
    positivity_tolerance=QUALIFICATION_TOLERANCE,
    edge_spectral_tolerance=QUALIFICATION_TOLERANCE,
    max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
)
loop_kwargs = dict(
    hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE,
    positivity_tolerance=QUALIFICATION_TOLERANCE,
)


def display_matrix(label, matrix, decimals=6):
    print(f"{label} =")
    print(np.round(matrix, decimals))
    print()


def build_reductions(eps_ab, eps_bc, eps_cd, eps_da, m_ab, m_bc, m_cd, m_da):
    rho = four_qubit_relational_loop_state(
        eps_ab, eps_bc, eps_cd, eps_da, m_ab, m_bc, m_cd, m_da, **core_kwargs
    )
    return rho, four_qubit_relational_loop_reductions(rho)


PRIMARY = dict(
    m_ab=IDENTITY2, m_bc=PAULI_X, m_cd=IDENTITY2, m_da=PAULI_Y,
    eps_ab=0.05, eps_bc=0.05, eps_cd=0.05, eps_da=0.05,
)

rho_primary, reductions_primary = build_reductions(
    PRIMARY["eps_ab"], PRIMARY["eps_bc"], PRIMARY["eps_cd"], PRIMARY["eps_da"],
    PRIMARY["m_ab"], PRIMARY["m_bc"], PRIMARY["m_cd"], PRIMARY["m_da"],
)

faithful_bound = 3.0 * (PRIMARY["eps_ab"] + PRIMARY["eps_bc"] + PRIMARY["eps_cd"] + PRIMARY["eps_da"])
print("faithful-domain LHS 3*sum(eps) =", faithful_bound, " (< 1 required)")
print("Tr(rho_ABCD)                   =", np.trace(rho_primary))
print("lambda_min(rho_ABCD)           =", np.min(np.linalg.eigvalsh(rho_primary)))
print("hermiticity residual           =", np.max(np.abs(rho_primary - rho_primary.conj().T)))
````

## 5. État global et réductions

Réductions de production (`four_qubit_relational_loop_reductions`) contre
oracles analytiques indépendants :

- arête : `rho_ij = (1-eps) I/4 + eps P_ij`
- site unique : `I/2`
- paire non-arête : `I/4`

````python
def phi_from_m(m_matrix):
    return (m_matrix / np.sqrt(2.0)).reshape(4)


def edge_oracle(m_matrix, eps):
    phi = phi_from_m(m_matrix)
    p_edge = np.outer(phi, phi.conj())
    return (1.0 - eps) * IDENTITY4 / 4.0 + eps * p_edge


oracle_ab = edge_oracle(PRIMARY["m_ab"], PRIMARY["eps_ab"])
oracle_bc = edge_oracle(PRIMARY["m_bc"], PRIMARY["eps_bc"])
oracle_cd = edge_oracle(PRIMARY["m_cd"], PRIMARY["eps_cd"])
oracle_da = edge_oracle(PRIMARY["m_da"], PRIMARY["eps_da"])

for label, computed, oracle in (
    ("rho_ab", reductions_primary["rho_ab"], oracle_ab),
    ("rho_bc", reductions_primary["rho_bc"], oracle_bc),
    ("rho_cd", reductions_primary["rho_cd"], oracle_cd),
    ("rho_da", reductions_primary["rho_da"], oracle_da),
):
    print(f"max|{label} - oracle| =", np.max(np.abs(computed - oracle)))

print()
for label in ("rho_a", "rho_b", "rho_c", "rho_d"):
    print(f"max|{label} - I/2| =", np.max(np.abs(reductions_primary[label] - IDENTITY2 / 2.0)))
print()
for label in ("rho_ac", "rho_bd"):
    print(f"max|{label} - I/4| =", np.max(np.abs(reductions_primary[label] - IDENTITY4 / 4.0)))
````

## 6. Pare-feu d'orientation `DA`

Section obligatoire. `rho_DA` est retourné en ordre `D tensor A`.

Fixture de régression d'orientation déterministe : `M_DA = R(pi/6)`,
autres `M = I`. On démontre que l'interprétation correcte `D tensor A`
et l'interprétation test-only incorrecte `A tensor D` (matrice
transposée) induisent des actions projectives différentes sur une
tangente de Pauli déterministe.

```text
CANONICAL_ORIENTATION = D_TENSOR_A
```

Aucune signification spatiale n'est attribuée à l'orientation.

````python
theta = np.pi / 6.0
m_da_rotation = np.array(
    [[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]], dtype=complex
)

rho_orient, reductions_orient = build_reductions(
    0.05, 0.05, 0.05, 0.05, IDENTITY2, IDENTITY2, IDENTITY2, m_da_rotation
)

print("rho_da shape =", reductions_orient["rho_da"].shape, " (D tensor A canonical order)")

oracle_da_orient = edge_oracle(m_da_rotation, 0.05)
print("max|rho_da - D(x)A oracle| =", np.max(np.abs(reductions_orient["rho_da"] - oracle_da_orient)))

# Correct D(x)A interpretation: the fixture matrix itself.
m_da_correct = m_da_rotation
# TEST_ONLY incorrect A(x)D interpretation: transposed matrix (this is
# exactly what an implicit/silent kron-order mistake would substitute).
m_da_incorrect = m_da_rotation.T

action_correct = apply_directional_link(
    m_da_correct, PAULI_X, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
)
action_incorrect = apply_directional_link(
    m_da_incorrect, PAULI_X, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
)

display_matrix("U_(D<-A)(X) [correct D(x)A]", action_correct)
display_matrix("U_(D<-A)(X) [TEST_ONLY incorrect A(x)D]", action_incorrect)
print("actions differ:", not np.allclose(action_correct, action_incorrect, atol=1e-6))

DA_AD_ORIENTATION_IS_PHYSICALLY_RELEVANT_TO_CONSTRUCTION = "YES_AT_MATRIX_CONTRACT_LEVEL"
CANONICAL_ORIENTATION = "D_TENSOR_A"
print()
print("DA_AD_ORIENTATION_IS_PHYSICALLY_RELEVANT_TO_CONSTRUCTION =",
      DA_AD_ORIENTATION_IS_PHYSICALLY_RELEVANT_TO_CONSTRUCTION)
print("CANONICAL_ORIENTATION =", CANONICAL_ORIENTATION)
````

## 7. Extraction modulaire d'arête

Pour chaque arête : `K_ij = -ln(rho_ij)` (production
`state_derived_edge_link`). Vérification indépendante :
`P_max(rho_ij) == P_min(K_ij)`.

```text
CORRELATION_MATRIX_SOURCE = MODULAR_GROUND_EIGENSPACE
```

````python
links_primary = {}
for edge in ("ab", "bc", "cd", "da"):
    rho_edge = reductions_primary[f"rho_{edge}"]
    link = state_derived_edge_link(rho_edge, **link_kwargs)
    links_primary[edge] = link

    w_rho, v_rho = np.linalg.eigh(rho_edge)
    w_k, v_k = np.linalg.eigh(link["modular_hamiltonian"])
    p_max_rho = np.outer(v_rho[:, -1], v_rho[:, -1].conj())
    p_min_k = np.outer(v_k[:, 0], v_k[:, 0].conj())

    print(f"--- edge {edge} ---")
    print("rho spectrum          =", w_rho)
    print("K spectrum            =", w_k)
    print("edge strength         =", link["strength"])
    print("max|P_max(rho)-P_min(K)| (independent) =", np.max(np.abs(p_max_rho - p_min_k)))
    print("max|link.modular_ground_projector - P_min(K)| =",
          np.max(np.abs(link["modular_ground_projector"] - p_min_k)))
    print()

CORRELATION_MATRIX_SOURCE = "MODULAR_GROUND_EIGENSPACE"
print("CORRELATION_MATRIX_SOURCE =", CORRELATION_MATRIX_SOURCE)
````

## 8. Force relationnelle d'arête

Oracle indépendant : `lambda_plus = (1+3 eps)/4`, `lambda_minus =
(1-eps)/4`, donc `lambda_plus - lambda_minus = eps`.

```text
EDGE_RELATIONAL_STRENGTH = STATE_DERIVED_SPECTRAL_GAP
EXTERNAL_EDGE_STRENGTH_INPUT_TO_RESPONSE = NONE
```

````python
for edge, eps in (("ab", PRIMARY["eps_ab"]), ("bc", PRIMARY["eps_bc"]),
                  ("cd", PRIMARY["eps_cd"]), ("da", PRIMARY["eps_da"])):
    lambda_plus = (1.0 + 3.0 * eps) / 4.0
    lambda_minus = (1.0 - eps) / 4.0
    oracle_strength = lambda_plus - lambda_minus
    extracted = links_primary[edge]["strength"]
    print(f"edge {edge}: oracle={oracle_strength:.10f}  extracted={extracted:.10f}  "
          f"|diff|={abs(oracle_strength - extracted):.2e}")

EDGE_RELATIONAL_STRENGTH = "STATE_DERIVED_SPECTRAL_GAP"
EXTERNAL_EDGE_STRENGTH_INPUT_TO_RESPONSE = "NONE"
print()
print("EDGE_RELATIONAL_STRENGTH                 =", EDGE_RELATIONAL_STRENGTH)
print("EXTERNAL_EDGE_STRENGTH_INPUT_TO_RESPONSE =", EDGE_RELATIONAL_STRENGTH and EXTERNAL_EDGE_STRENGTH_INPUT_TO_RESPONSE)
````

## 9. Lien directionnel

`correlation_matrix` extrait est montré uniquement comme représentant
(aucune comparaison brute à la fixture sans quotient de phase — la
comparaison se fait au niveau projecteur, §7). Démonstration de
`U_(i<-j)(X) = M X^T M^dagger` sur `X, Y, Z` : hermiticité, absence de
trace, préservation de la norme de Hilbert-Schmidt.

````python
for edge in ("ab", "bc", "cd", "da"):
    m_matrix = links_primary[edge]["correlation_matrix"]
    display_matrix(f"M_{edge} (representative only, phase-gauge)", m_matrix)

for edge in ("ab", "bc", "cd", "da"):
    m_matrix = links_primary[edge]["correlation_matrix"]
    print(f"--- edge {edge} ---")
    for name, xi in (("X", PAULI_X), ("Y", PAULI_Y), ("Z", PAULI_Z)):
        out = apply_directional_link(
            m_matrix, xi, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
            trace_tolerance=QUALIFICATION_TOLERANCE,
            max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
        )
        herm_resid = np.max(np.abs(out - out.conj().T))
        trace_resid = abs(np.trace(out))
        hs_in = np.sqrt(np.trace(xi.conj().T @ xi).real)
        hs_out = np.sqrt(np.trace(out.conj().T @ out).real)
        print(f"  {name}: hermiticity_resid={herm_resid:.2e}  trace_resid={trace_resid:.2e}  "
              f"HS_in={hs_in:.6f}  HS_out={hs_out:.6f}")
    print()
````

## 10. Pare-feu de phase

Application de phases déterministes indépendantes `M_ij -> e^{i
theta_ij} M_ij`. `P_ij` et `U_(i<-j)` restent inchangés.

```text
RAW_M_PHASE = GAUGE
DIRECTIONAL_LINK_ACTION = PHASE_INDEPENDENT
```

````python
independent_thetas = dict(ab=0.31, bc=0.77, cd=1.19, da=2.03)

for edge, theta_edge in independent_thetas.items():
    m_matrix = links_primary[edge]["correlation_matrix"]
    m_phased = np.exp(1j * theta_edge) * m_matrix
    print(f"--- edge {edge} (theta={theta_edge}) ---")
    print("raw M changed:", not np.allclose(m_matrix, m_phased, atol=1e-10))

    phi = phi_from_m(m_matrix)
    phi_phased = phi_from_m(m_phased)
    p_original = np.outer(phi, phi.conj())
    p_phased = np.outer(phi_phased, phi_phased.conj())
    print("max|P_ij - P_ij(phased)| =", np.max(np.abs(p_original - p_phased)))

    for name, xi in (("X", PAULI_X), ("Y", PAULI_Y), ("Z", PAULI_Z)):
        out_original = apply_directional_link(
            m_matrix, xi, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
            trace_tolerance=QUALIFICATION_TOLERANCE,
            max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
        )
        out_phased = apply_directional_link(
            m_phased, xi, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
            trace_tolerance=QUALIFICATION_TOLERANCE,
            max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
        )
        print(f"  max|U({name}) - U_phased({name})| =", np.max(np.abs(out_original - out_phased)))
    print()

RAW_M_PHASE = "GAUGE"
DIRECTIONAL_LINK_ACTION = "PHASE_INDEPENDENT"
print("RAW_M_PHASE              =", RAW_M_PHASE)
print("DIRECTIONAL_LINK_ACTION  =", DIRECTIONAL_LINK_ACTION)
````

## 11. Lien inverse

Production `reverse_correlation_matrix`. `M_ji = M_ij^T`.
`U_(j<-i) o U_(i<-j) = identité` pour des tangentes déterministes.

Les transferts d'arête centrés PONDÉRÉS ne sont PAS des inverses :

```text
CENTERED_WEIGHTED_EDGE_TRANSFERS_ARE_INVERSES = NO
```

````python
edge = "bc"
m_bc = links_primary[edge]["correlation_matrix"]
strength_bc = links_primary[edge]["strength"]
m_cb = reverse_correlation_matrix(m_bc, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE)
print("max|M_cb - M_bc^T| =", np.max(np.abs(m_cb - m_bc.T)))

for name, xi in (("X", PAULI_X), ("Y", PAULI_Y), ("Z", PAULI_Z)):
    forward = apply_directional_link(
        m_bc, xi, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
        trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
    )
    back = apply_directional_link(
        m_cb, forward, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
        trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
    )
    print(f"U_(j<-i) o U_(i<-j)({name}): max|back - {name}| =", np.max(np.abs(back - xi)))

    # Weighted (centered) transfer composition uses the SAME edge strength
    # in both directions: L_(j<-i) o L_(i<-j)(X) = strength^2 * X != X.
    forward_weighted = strength_bc * forward
    back_weighted = strength_bc * apply_directional_link(
        m_cb, forward_weighted, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
        trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
    )
    print(f"  weighted L_(j<-i) o L_(i<-j)({name}): max|result - {name}| =",
          np.max(np.abs(back_weighted - xi)), f" (strength^2={strength_bc**2:.6f})")

CENTERED_WEIGHTED_EDGE_TRANSFERS_ARE_INVERSES = "NO"
print()
print("CENTERED_WEIGHTED_EDGE_TRANSFERS_ARE_INVERSES =", CENTERED_WEIGHTED_EDGE_TRANSFERS_ARE_INVERSES)
````

```text
CENTERED_EDGE_TRANSFER = STATE_DERIVED
NORMALIZATION_REPAIR = NONE
```

````python
for edge, eps in (("ab", PRIMARY["eps_ab"]), ("bc", PRIMARY["eps_bc"]),
                  ("cd", PRIMARY["eps_cd"]), ("da", PRIMARY["eps_da"])):
    rho_edge = reductions_primary[f"rho_{edge}"]
    m_matrix = links_primary[edge]["correlation_matrix"]
    print(f"--- edge {edge} ---")
    for name, xi in (("X", PAULI_X), ("Y", PAULI_Y), ("Z", PAULI_Z)):
        transferred = state_derived_centered_edge_transfer(
            rho_edge, xi, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
            trace_tolerance=QUALIFICATION_TOLERANCE, positivity_tolerance=QUALIFICATION_TOLERANCE,
        )
        action = apply_directional_link(
            m_matrix, xi, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
            trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
        )
        residual = np.max(np.abs(transferred - eps * action))
        print(f"  {name}: max|L - eps*U| = {residual:.2e}")
    print()

CENTERED_EDGE_TRANSFER = "STATE_DERIVED"
NORMALIZATION_REPAIR = "NONE"
print("CENTERED_EDGE_TRANSFER =", CENTERED_EDGE_TRANSFER)
print("NORMALIZATION_REPAIR   =", NORMALIZATION_REPAIR)
````

## 13. Holonomie de boucle brute

Production `projective_loop_holonomy`. Oracle de la fixture primaire :
`H_A = -i sigma_Z` à phase scalaire près.

```text
RAW_HOLONOMY_MATRIX = DEFINED_UP_TO_GLOBAL_PHASE
```

## 23. Covariance de base locale

Repères déterministes non triviaux `V_A, V_B, V_C, V_D`. La chaîne
complète (état global -> réductions -> logique SWAP/orientation `DA`
-> liens -> réponse) est repassée sous transformation. `R_square'(V_A
X V_A^dagger) = V_A R_square(X) V_A^dagger`.

````python
def fixed_unitary_2_phase(theta_value, phi_value):
    c, s = np.cos(theta_value), np.sin(theta_value)
    rotation = np.array([[c, -s], [s, c]], dtype=complex)
    phase = np.diag([1.0, np.exp(1j * phi_value)])
    return rotation @ phase


v_a = fixed_unitary_2_phase(0.61, 0.29)
v_b = fixed_unitary_2_phase(0.37, 0.53)
v_c = fixed_unitary_2_phase(0.44, 0.17)
v_d = fixed_unitary_2_phase(0.71, 1.02)

v_full = np.kron(np.kron(np.kron(v_a, v_b), v_c), v_d)
rho_prime = v_full @ rho_primary @ v_full.conj().T
reductions_prime = four_qubit_relational_loop_reductions(rho_prime)

xi = PAULI_X
xi_prime = v_a @ xi @ v_a.conj().T

r_original = relational_curvature_response_candidate(
    reductions_primary["rho_ab"], reductions_primary["rho_bc"],
    reductions_primary["rho_cd"], reductions_primary["rho_da"], xi, **link_kwargs
)
r_prime = relational_curvature_response_candidate(
    reductions_prime["rho_ab"], reductions_prime["rho_bc"],
    reductions_prime["rho_cd"], reductions_prime["rho_da"], xi_prime, **link_kwargs
)
expected = v_a @ r_original @ v_a.conj().T
residual = np.max(np.abs(r_prime - expected))
print("max|R_square'(V_A X V_A^dagger) - V_A R_square(X) V_A^dagger| =", residual)

FRAME_FIREWALL = "PASS_QUALIFICATION_FOR_DECLARED_FIXTURE"
print()
print("FRAME_FIREWALL =", FRAME_FIREWALL, " (raw M/H matrices need not be invariant)")
````

````python
params_f8 = dict(eps_ab=0.04, eps_bc=0.05, eps_cd=0.03, eps_da=0.06,
                  m_ab=m_ab_gauge, m_bc=m_bc_gauge, m_cd=m_cd_gauge, m_da=m_da_gauge)
_, reductions_f8 = build_reductions(
    params_f8["eps_ab"], params_f8["eps_bc"], params_f8["eps_cd"], params_f8["eps_da"],
    params_f8["m_ab"], params_f8["m_bc"], params_f8["m_cd"], params_f8["m_da"],
)
links_f8 = {e: state_derived_edge_link(reductions_f8[f"rho_{e}"], **link_kwargs)
            for e in ("ab", "bc", "cd", "da")}

strength_ab_bc = links_f8["ab"]["strength"] * links_f8["bc"]["strength"]
strength_da_cd = links_f8["da"]["strength"] * links_f8["cd"]["strength"]
print("strength_AB*strength_BC =", strength_ab_bc)
print("strength_DA*strength_CD =", strength_da_cd)
print("unequal path amplitudes:", not np.isclose(strength_ab_bc, strength_da_cd))

# TEST-ONLY open path A-B-C: L_(A<-B) o L_(B<-C), starting from a tangent at C.
tangent_c = PAULI_X
m_bc_f8 = links_f8["bc"]["correlation_matrix"]
m_ab_f8 = links_f8["ab"]["correlation_matrix"]
step1_bc = params_f8["eps_bc"] * apply_directional_link(
    m_bc_f8, tangent_c, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
)
path_abc = params_f8["eps_ab"] * apply_directional_link(
    m_ab_f8, step1_bc, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
)

# TEST-ONLY open path A-D-C: L_(A<-D) o L_(D<-C), using reverse-link contract
# on the stored CD and DA edges (spec §12: M_ji = M_ij^T, same edge strength).
m_cd_f8 = links_f8["cd"]["correlation_matrix"]
m_da_f8 = links_f8["da"]["correlation_matrix"]
m_dc_f8 = reverse_correlation_matrix(m_cd_f8, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE)
m_ad_f8 = reverse_correlation_matrix(m_da_f8, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE)
step1_dc = params_f8["eps_cd"] * apply_directional_link(
    m_dc_f8, tangent_c, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
)
path_adc = params_f8["eps_da"] * apply_directional_link(
    m_ad_f8, step1_dc, hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE, max_entanglement_unitarity_tolerance=UNITARITY_TOLERANCE,
)

display_matrix("open path A-B-C (test-only, unnormalized)", path_abc)
display_matrix("open path A-D-C (test-only, unnormalized)", path_adc)
print("open-path outputs differ:", not np.allclose(path_abc, path_adc, atol=1e-6))

# Closed loop: mandatory invariant.
for name, xi in (("X", PAULI_X), ("Y", PAULI_Y), ("Z", PAULI_Z)):
    r_f8 = relational_curvature_response_candidate(
        reductions_f8["rho_ab"], reductions_f8["rho_bc"], reductions_f8["rho_cd"], reductions_f8["rho_da"],
        xi, **link_kwargs
    )
    print(f"closed loop {name}: max|R_square| =", np.max(np.abs(r_f8)))

OPEN_PATH_RAW_DIFFERENCE = "NOT_CURVATURE_INVARIANT"
UNEQUAL_PATH_ATTENUATION = "NOT_CURVATURE_BY_ITSELF"
F8 = "PASS_NEGATIVE_CONTROL"
print()
print("OPEN_PATH_RAW_DIFFERENCE =", OPEN_PATH_RAW_DIFFERENCE)
print("UNEQUAL_PATH_ATTENUATION =", UNEQUAL_PATH_ATTENUATION)
print("F8                       =", F8)
````
