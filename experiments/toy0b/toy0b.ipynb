## 2. Le système `A|B|C`

$$
\mathcal H_{ABC} = \mathcal H_A \otimes \mathcal H_B \otimes \mathcal H_C,
\qquad
\mathcal H_A = \mathcal H_B = \mathcal H_C = \mathbb C^2
$$

(`docs/toy-models/toy0b/specification.md` §4). Schématiquement :

```text
AB ---- B ---- BC
```

`B` est le sous-système **commun** aux deux relations `AB` et `BC` : un
chevauchement **algébrique** déclaré de la factorisation tensorielle
`A|B|C`, pas un point spatial ni une position géométrique.

## 3. Famille d'états `rho_ABC(beta, lambda_, mu)`

$$
\rho_{ABC} = \frac18\Big[I + \beta Z_B + \lambda X_A X_B + \mu Y_B Y_C\Big]
$$

(`docs/toy-models/toy0b/specification.md` §5), construite en appelant
directement `three_qubit_overlapping_pauli_relation_state` (aucune
construction manuelle de matrice dans ce notebook). Le domaine fidèle
exact (§6) est :

$$
\beta^2 + \lambda^2 + \mu^2 < 1.
$$

On affiche ci-dessous la matrice du cas principal et son rayon `r`
uniquement comme contrôle de domaine (`r < 1`), sans introduire aucun
seuil ni score supplémentaire.

````python
# Helper purement présentationnel : arrondi d'affichage d'une matrice.
# Ne transforme aucune donnée scientifique ; sert uniquement à la lecture.
def display_matrix(label, matrix, decimals=4):
    print(f"{label} =")
    print(np.round(matrix, decimals))
    print()


rho = {
    name: three_qubit_overlapping_pauli_relation_state(p["beta"], p["lambda_"], p["mu"])
    for name, p in fixture_params.items()
}

display_matrix("rho_ABC [MAIN]", rho["MAIN"])

r_main = np.sqrt(
    fixture_params["MAIN"]["beta"] ** 2
    + fixture_params["MAIN"]["lambda_"] ** 2
    + fixture_params["MAIN"]["mu"] ** 2
)
print(f"r [MAIN] = {r_main}  (contrôle de domaine, r < 1 : {r_main < 1.0})")
print(f"Tr(rho_ABC) [MAIN] = {np.trace(rho['MAIN'])}")
````

## 4. États réduits

On extrait `rho_AB`, `rho_BC`, `rho_B`, `rho_A`, `rho_C` du cas principal
via `cosmotgg.core.states.partial_trace` (`docs/toy-models/toy0b/specification.md`
§7). `rho_AB` porte la relation `A-B` (paramétrée par `lambda`) ; `rho_BC`
porte la relation `B-C` (paramétrée par `mu`) : ce sont **deux relations
distinctes**, qui partagent néanmoins le même facteur commun `rho_B`
(paramétré par `beta` seul).

````python
from cosmotgg.core.states import partial_trace

rho_ab = partial_trace(rho["MAIN"], dimensions=(2, 2, 2), keep=[0, 1])
rho_bc = partial_trace(rho["MAIN"], dimensions=(2, 2, 2), keep=[1, 2])
rho_b_from_ab = partial_trace(rho_ab, dimensions=(2, 2), keep=[1])
rho_b_from_bc = partial_trace(rho_bc, dimensions=(2, 2), keep=[0])
rho_b = partial_trace(rho["MAIN"], dimensions=(2, 2, 2), keep=[1])
rho_a = partial_trace(rho["MAIN"], dimensions=(2, 2, 2), keep=[0])
rho_c = partial_trace(rho["MAIN"], dimensions=(2, 2, 2), keep=[2])

display_matrix("rho_AB [MAIN]", rho_ab)
display_matrix("rho_BC [MAIN]", rho_bc)
display_matrix("rho_B [MAIN]", rho_b)
display_matrix("rho_A [MAIN]", rho_a)
display_matrix("rho_C [MAIN]", rho_c)

print(
    "rho_B (via rho_AB) == rho_B (via rho_BC) : "
    f"{np.allclose(rho_b_from_ab, rho_b_from_bc)}"
)
print(f"rho_B (via rho_AB) == rho_B (trace directe) : {np.allclose(rho_b_from_ab, rho_b)}")
print(f"rho_AB et rho_BC sont deux relations distinctes (formes différentes, même B partagé).")
````

## 5. Structures modulaires `K_AB`, `K_BC`

$$
K_{AB} = -\ln(\rho_{AB}), \qquad K_{BC} = -\ln(\rho_{BC})
$$

(`docs/toy-models/toy0b/specification.md` §8), via
`cosmotgg.core.modular.modular_hamiltonian`. Aucun paramètre modulaire
`s` n'est introduit ici : ce sont deux hamiltoniens modulaires
statiques.

Ces deux structures vivent cependant sur deux algèbres différentes,
`B(H_A ⊗ H_B)` et `B(H_B ⊗ H_C)` : elles ne sont donc pas directement
comparables terme à terme. Comment les comparer malgré cela ?

````python
from cosmotgg.core.modular import modular_hamiltonian

k_ab = modular_hamiltonian(rho_ab, **core_tolerances)
k_bc = modular_hamiltonian(rho_bc, **core_tolerances)

display_matrix("K_AB [MAIN]", k_ab)
display_matrix("K_BC [MAIN]", k_bc)
````

## 6. Le chevauchement `B`

La solution contractuelle (`docs/toy-models/toy0b/specification.md` §9)
consiste à projeter chaque hamiltonien modulaire sur l'algèbre commune
`B` via la conditional expectation **traciale** de ce toy type-I fini :

$$
E_B^A(X_{AB}) = \frac{\operatorname{Tr}_A(X_{AB})}{2},
\qquad
E_B^C(X_{BC}) = \frac{\operatorname{Tr}_C(X_{BC})}{2}.
$$

Cette cellule ne définit aucune fonction scientifique locale : elle
utilise explicitement `partial_trace(...)/2`, exactement comme déjà
spécifié (§9 de la spécification) et déjà implémenté/testé dans
`overlap_relative_modular_generator`.

````python
e_a_k_ab = partial_trace(k_ab, dimensions=(2, 2), keep=[1]) / 2
e_c_k_bc = partial_trace(k_bc, dimensions=(2, 2), keep=[0]) / 2

display_matrix("E_B^A(K_AB) [MAIN]", e_a_k_ab)
display_matrix("E_B^C(K_BC) [MAIN]", e_c_k_bc)
````

## 7. Réduction sans trace et `Delta_B`

$$
\operatorname{tl}_B(X) = X - \frac{\operatorname{Tr}(X)}{2} I_B,
\qquad
\Delta_{(A:C|B)} = -\operatorname{tl}_B\big[E_B^A(K_{AB}) - E_B^C(K_{BC})\big]
$$

(§11–§12 de la spécification). Pour le **résultat scientifique
officiel**, on utilise directement le code de production testé
`overlap_relative_modular_generator` — pas la reconstruction manuelle de
la section précédente, qui n'était qu'illustrative des ingrédients :

````python
from cosmotgg.models.model0b.relative import overlap_relative_modular_generator

delta_b = {
    name: overlap_relative_modular_generator(rho[name], **core_tolerances)
    for name in rho
}

display_matrix("Delta_B [MAIN]", delta_b["MAIN"])
print(f"Tr(Delta_B) [MAIN]       = {np.trace(delta_b['MAIN'])}")
print(f"Delta_B hermitien [MAIN] = {np.allclose(delta_b['MAIN'], delta_b['MAIN'].conj().T)}")
````

**Point central** : `Delta_B` ne dépend d'aucun paramètre modulaire
partagé `s` — c'est un objet déterminé **directement** par l'état
`rho_ABC(beta, lambda_, mu)`, sans flot fini, sans exponentielle.

## 8. Contrôle analytique

Identité analytique déjà enregistrée et testée
(`docs/toy-models/toy0b/specification.md` §14) :

$$
\Delta_{(A:C|B)} = \beta\big[f(r_{AB}) - f(r_{BC})\big] Z_B,
\qquad
f(r) = \frac{\operatorname{atanh}(r)}{r},
\qquad
r_{AB} = \sqrt{\beta^2+\lambda^2},
\quad
r_{BC} = \sqrt{\beta^2+\mu^2}.
$$

On calcule ci-dessous cet oracle **directement depuis cette formule**
(aucune fonction scientifique helper : l'expression est écrite en ligne),
pour le cas principal, et on compare au `Delta_B` obtenu par la
construction modulaire de la section précédente.

````python
beta_main, lambda_main, mu_main = (
    fixture_params["MAIN"]["beta"],
    fixture_params["MAIN"]["lambda_"],
    fixture_params["MAIN"]["mu"],
)

r_ab_main = np.hypot(beta_main, lambda_main)
r_bc_main = np.hypot(beta_main, mu_main)

pauli_z = np.array([[1.0, 0.0], [0.0, -1.0]], dtype=complex)

delta_analytic_main = (
    beta_main
    * (np.arctanh(r_ab_main) / r_ab_main - np.arctanh(r_bc_main) / r_bc_main)
    * pauli_z
)

delta_modular_main = delta_b["MAIN"]

display_matrix("Delta_modular [MAIN]", delta_modular_main)
display_matrix("Delta_analytic [MAIN]", delta_analytic_main)

residual_main = np.max(np.abs(delta_modular_main - delta_analytic_main))
print(
    f"max|Delta_modular - Delta_analytic| [MAIN] = {residual_main}"
    "  (NON_NORMATIVE_QUALIFICATION_NUMERICS)"
)
````

Fixture `R1` : `|lambda| = |mu| != 0` (`lambda=0.3`, `mu=-0.3`). On
vérifie d'abord, **indépendamment** de `Delta_B`, que les deux relations
`AB` et `BC` sont chacune non triviales :

````python
rho_r1 = rho["R1"]
rho_ab_r1 = partial_trace(rho_r1, dimensions=(2, 2, 2), keep=[0, 1])
rho_bc_r1 = partial_trace(rho_r1, dimensions=(2, 2, 2), keep=[1, 2])
rho_a_r1 = partial_trace(rho_r1, dimensions=(2, 2, 2), keep=[0])
rho_b_r1 = partial_trace(rho_r1, dimensions=(2, 2, 2), keep=[1])
rho_c_r1 = partial_trace(rho_r1, dimensions=(2, 2, 2), keep=[2])

print(
    "rho_AB != rho_A (x) rho_B [R1] : "
    f"{not np.allclose(rho_ab_r1, np.kron(rho_a_r1, rho_b_r1))}"
)
print(
    "rho_BC != rho_B (x) rho_C [R1] : "
    f"{not np.allclose(rho_bc_r1, np.kron(rho_b_r1, rho_c_r1))}"
)

display_matrix("Delta_B [R1]", delta_b["R1"])
print(f"Delta_B ~= 0 [R1] : {np.allclose(delta_b['R1'], np.zeros((2, 2)))}")
````

## 15. Covariance locale

Contrôle : sous `U = U_A ⊗ U_B ⊗ U_C` (déterministe,
`NON_NORMATIVE_QUALIFICATION_FIXTURE`), on vérifie

$$
\Delta(\rho') = U_B\,\Delta(\rho)\,U_B^\dagger,
\qquad
\rho' = U\,\rho\,U^\dagger
$$

à précision machine (`docs/toy-models/toy0b/specification.md` §21).

**Frontière explicite** :

```text
LOCAL_PRODUCT_UNITARY_COVARIANCE != GLOBAL_UNITARY_COVARIANCE
LOCAL_PRODUCT_UNITARY_COVARIANCE != REFACTORIZATION_INVARIANCE
LOCAL_PRODUCT_UNITARY_COVARIANCE != LORENTZ_OR_GENERAL_COVARIANCE
```

````python
# NON_NORMATIVE_QUALIFICATION_FIXTURE : unitaires locaux déterministes,
# sans statut physique privilégié, choisis pour sortir de la base
# canonique.
sqrt2 = np.sqrt(2.0)
U_A = (1.0 / sqrt2) * np.array([[1.0, 1.0], [1.0, -1.0]], dtype=complex)
U_B = (1.0 / sqrt2) * np.array([[1.0, 1.0j], [1.0j, 1.0]], dtype=complex)
U_C = np.array([[1.0, 0.0], [0.0, 1.0j]], dtype=complex)
U_ABC = np.kron(U_A, np.kron(U_B, U_C))

print("max|U_A^dagger U_A - I| =", np.max(np.abs(U_A.conj().T @ U_A - np.eye(2))))
print("max|U_B^dagger U_B - I| =", np.max(np.abs(U_B.conj().T @ U_B - np.eye(2))))
print("max|U_C^dagger U_C - I| =", np.max(np.abs(U_C.conj().T @ U_C - np.eye(2))))
print("max|U^dagger U - I|     =", np.max(np.abs(U_ABC.conj().T @ U_ABC - np.eye(8))))

rho_main_transformed = U_ABC @ rho["MAIN"] @ U_ABC.conj().T
delta_b_transformed = overlap_relative_modular_generator(rho_main_transformed, **core_tolerances)
expected_transformed = U_B @ delta_b["MAIN"] @ U_B.conj().T

residual_covariance = np.max(np.abs(delta_b_transformed - expected_transformed))
print(
    "max|Delta(rho') - U_B Delta(rho) U_B^dagger| = "
    f"{residual_covariance}  (NON_NORMATIVE_QUALIFICATION_NUMERICS)"
)
````

## 16. Limitation de colinéarité (section obligatoire)

```text
OVERLAP_DERIVATION_COLLINEAR_WITH_LOCAL_MODULAR_GENERATOR =
    TRUE_FOR_DECLARED_STATE_FAMILY
```

(`docs/toy-models/toy0b/specification.md` §18). Sur toute la famille
déclarée, `Delta_B` reste colinéaire à `tl_B(ln(rho_B))`, la partie sans
trace du logarithme de `rho_B` seul. On l'affiche ci-dessous pour le cas
`MAIN` (aucun ratio, norme ou score normatif n'est créé) :

````python
from cosmotgg.core.modular import hermitian_log

log_rho_b_main = hermitian_log(
    rho_b,
    hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    positivity_tolerance=QUALIFICATION_TOLERANCE,
)
tl_log_rho_b_main = log_rho_b_main - (np.trace(log_rho_b_main) / 2.0) * np.eye(2)

display_matrix("Delta_B [MAIN]", delta_b["MAIN"])
display_matrix("tl_B(ln(rho_B)) [MAIN]", tl_log_rho_b_main)
````

Dans cette famille déclarée, `Delta_B` n'apporte donc pas une nouvelle
direction opératorielle sur `B(H_B)` par rapport à la structure modulaire
locale déjà portée par `rho_B` seul : l'information relationnelle
supplémentaire apportée par le chevauchement `AB`/`BC` est essentiellement
portée par l'**amplitude** `delta`, pas par une nouvelle direction. C'est
une limite du candidat pour cette famille d'états, pas un résultat
caché.

## 18. Pourquoi T1 n'est toujours pas passé

On pourrait mathématiquement envisager d'utiliser `Delta_B` comme
générateur d'un flot fini, en écrivant formellement :

```text

```

**Cette expression n'est ni exécutée ni construite dans ce notebook** :
aucune fonction d'exponentielle matricielle n'est importée ni appelée à
cette fin.

Le problème reste entier même formellement : `tau` n'est déterminé par
**aucune donnée de l'état** `rho_ABC` — ni par `Delta_B`, ni par la
famille d'états, ni par aucun autre objet déjà défini dans ce notebook.

```text

```

Conclusion : `Delta_B` fixe un générateur algébrique relatif — pas une
durée, pas une horloge, pas un changement physique fini.
