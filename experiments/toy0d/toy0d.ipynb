## 3. État réduit réel du chevauchement `rho_B`

`rho_B` est l'état réduit **physique** du sous-système de chevauchement
`B`, calculé via `cosmotgg.core.states.partial_trace`. Ni `omega_A` ni
`omega_C` (§4 ci-dessous) ne seront jamais appelés « état physique de
`B` ».

````python
from cosmotgg.core.states import partial_trace

rho_b = partial_trace(rho_abc, dimensions=(2, 2, 2), keep=[1])
identity2 = np.eye(2, dtype=complex)

display_matrix("rho_B", rho_b)
print("rho_B == I/2 :", np.allclose(rho_b, identity2 / 2, atol=1e-10))
````

## 4. Reconstruction des états contextuels `omega_A`, `omega_C`

Via `cosmotgg.models.model0d.transport.contextual_state_from_projected_generator`
(`docs/toy-models/toy0d/specification.md` §5). `omega_A`, `omega_C` sont
des états contextuels auxiliaires, reconstruits à partir de `chi_A`,
`chi_C` — jamais des états physiques de `B`.

````python
from cosmotgg.models.model0d.transport import contextual_state_from_projected_generator

omega_a = contextual_state_from_projected_generator(chi_a, **chi_tolerances)
omega_c = contextual_state_from_projected_generator(chi_c, **chi_tolerances)

display_matrix("omega_A", omega_a)
display_matrix("omega_C", omega_c)

print("Tr(omega_A) =", np.trace(omega_a))
print("Tr(omega_C) =", np.trace(omega_c))
print("eigvals(omega_A) > 0 :", np.all(np.linalg.eigvalsh(omega_a) > 0.0))
print("eigvals(omega_C) > 0 :", np.all(np.linalg.eigvalsh(omega_c) > 0.0))
print()
print("omega_A != rho_B  :", not np.allclose(omega_a, rho_b, atol=1e-6))
print("omega_C != rho_B  :", not np.allclose(omega_c, rho_b, atol=1e-6))
print("omega_A != omega_C:", not np.allclose(omega_a, omega_c, atol=1e-6))
print()
print("CONTEXTUAL_STATE_PHYSICAL_STATUS = AUXILIARY_STATE_RECONSTRUCTED_FROM_PROJECTED_MODULAR_DATA")
````

## 5. Invariance à la partie scalaire de `chi`

Deux scalaires d'affichage non normatifs (`c`, `c'`) sont ajoutés à
`chi_A`, `chi_C` ; la reconstruction `omega` doit être algébriquement
inchangée sous normalisation (§5-§6 de la spécification).

````python
# NON_NORMATIVE_QUALIFICATION_FIXTURE : scalaires d'affichage arbitraires.
DISPLAY_SCALAR_C = 1.3
DISPLAY_SCALAR_C_PRIME = -0.7

omega_a_shifted = contextual_state_from_projected_generator(
    chi_a + DISPLAY_SCALAR_C * identity2, **chi_tolerances
)
omega_c_shifted = contextual_state_from_projected_generator(
    chi_c + DISPLAY_SCALAR_C_PRIME * identity2, **chi_tolerances
)

print("max|omega_A(chi_A + c I) - omega_A(chi_A)|   =", np.max(np.abs(omega_a_shifted - omega_a)))
print("max|omega_C(chi_C + c' I) - omega_C(chi_C)|  =", np.max(np.abs(omega_c_shifted - omega_c)))
print()
print("ADDITIVE_IDENTITY_AMBIGUITY = REMOVED_BY_NORMALIZATION")
print("PARAMETER_FREE_RECONSTRUCTION_IS_INHERITED_FROM_K_CONVENTION = TRUE")
````

## 6. Oracle indépendant des états contextuels (`d_B = 2`)

Pour cette famille qubit canonique, `chi_A = a X_B` et `chi_C = c Y_B`
exactement (résultat déjà qualifié par `model0c`, §11 de sa
spécification). Le calcul fonctionnel `exp(-a X_B)/Tr(...)` admet la
forme close `omega_A = 1/2 [I - tanh(a) X_B]` (et de même pour `omega_C`
avec `Y_B`), construite ici en ligne, indépendamment du chemin spectral
de production.

````python
import math

pauli_x = np.array([[0.0, 1.0], [1.0, 0.0]], dtype=complex)
pauli_y = np.array([[0.0, -1.0j], [1.0j, 0.0]], dtype=complex)
pauli_z = np.array([[1.0, 0.0], [0.0, -1.0]], dtype=complex)

a_scalar = (np.trace(pauli_x @ chi_a) / 2.0).real
c_scalar = (np.trace(pauli_y @ chi_c) / 2.0).real
print("a_scalar (chi_A = a_scalar * X_B) =", a_scalar)
print("c_scalar (chi_C = c_scalar * Y_B) =", c_scalar)
print("chi_A == a_scalar * X_B :", np.allclose(chi_a, a_scalar * pauli_x, atol=1e-8))
print("chi_C == c_scalar * Y_B :", np.allclose(chi_c, c_scalar * pauli_y, atol=1e-8))
print()

omega_a_analytic = 0.5 * (identity2 - math.tanh(a_scalar) * pauli_x)
omega_c_analytic = 0.5 * (identity2 - math.tanh(c_scalar) * pauli_y)

print("max|omega_A_production - omega_A_analytic| =", np.max(np.abs(omega_a - omega_a_analytic)))
print("max|omega_C_production - omega_C_analytic| =", np.max(np.abs(omega_c - omega_c_analytic)))
print()
print("ANALYTIC_D2_ORACLE = NON_NORMATIVE_QUALIFICATION_ORACLE")
````

## 8. Oracle indépendant de `F` (calcul spectral, `d_B = 2`)

Oracle construit en ligne par diagonalisation directe
(`numpy.linalg.eigh`), sans appeler `connes_cocycle_at_minus_i_half`.

````python
eigvals_a, eigvecs_a = np.linalg.eigh(omega_a)
eigvals_c, eigvecs_c = np.linalg.eigh(omega_c)

sqrt_omega_c = (eigvecs_c * np.sqrt(eigvals_c)) @ eigvecs_c.conj().T
invsqrt_omega_a = (eigvecs_a * (eigvals_a ** -0.5)) @ eigvecs_a.conj().T
f_oracle = sqrt_omega_c @ invsqrt_omega_a

display_matrix("F_production", f_ac)
display_matrix("F_oracle (sqrt(omega_C) @ invsqrt(omega_A), spectral, independent)", f_oracle)

analytic_oracle_residual = np.max(np.abs(f_ac - f_oracle))
print("analytic_oracle_residual =", analytic_oracle_residual)
````

## 9. Transport exact

````python
transported = f_ac @ omega_a @ f_ac.conj().T
display_matrix("F_AC omega_A F_AC^dagger", transported)
display_matrix("omega_C", omega_c)

transport_residual = np.max(np.abs(transported - omega_c))
print("transport_residual =", transport_residual)
print()
print("EXACT_CONTEXTUAL_STATE_TRANSPORT = PASS_QUALIFICATION")
print()
print("THIS_IS_NOT_A_PHYSICAL_STATE_EVOLUTION")
````

## 10. Identité et inverse

Aucun vocabulaire de réversibilité temporelle n'est employé ci-dessous :
`F_AA` et `F_CA F_AC` sont des identités algébriques exactes, pas des
énoncés sur un ordre physique.

````python
f_aa = finite_relative_contextual_state_transporter(omega_a, omega_a, **core_tolerances)
f_ca = finite_relative_contextual_state_transporter(omega_c, omega_a, **core_tolerances)

print("max|F_AA - I|      =", np.max(np.abs(f_aa - identity2)))
print("max|F_CA F_AC - I| =", np.max(np.abs(f_ca @ f_ac - identity2)))
````

## 11. Décomposition polaire de `F_AC`

`F = U P` (`docs/toy-models/toy0d/specification.md` §12, formule déjà
figée, calculée ici par SVD) : `U` capture l'orientation/la
non-commutation de la paire, `spectrum(P)` capture le désaccord
spectral/d'amplitude. `U` n'est **pas** la phase d'Uhlmann (§12).

````python
def polar_decomposition(f_matrix):
    """Décomposition polaire droite F = U P (formule déjà figée, spec §12)."""
    w_matrix, singular_values, vh_matrix = np.linalg.svd(f_matrix)
    u_matrix = w_matrix @ vh_matrix
    p_matrix = vh_matrix.conj().T @ np.diag(singular_values) @ vh_matrix
    return u_matrix, p_matrix


u_ac, p_ac = polar_decomposition(f_ac)
display_matrix("U (F_AC)", u_ac)
display_matrix("P (F_AC)", p_ac)
print("||F_AC - U P|| =", np.max(np.abs(f_ac - u_ac @ p_ac)))
print()
print("U != I (fixture C3)  :", not np.allclose(u_ac, identity2, atol=1e-6))
print("N != 0                :", not np.allclose(n_matrix, zero_2x2, atol=1e-8))
print()
print("POLAR_NONCOMMUTING_COMPONENT_ALIGNS_WITH_MODEL0C_N_REGIME = PASS_FOR_DECLARED_UPSTREAM_FIXTURE")
print("U IS NOT UHLMANN_PHASE")
````

## 16. `D4` — non-canalité

````python
f_dagger_f = f_ac.conj().T @ f_ac
display_matrix("F_AC^dagger F_AC", f_dagger_f)
print("max|F^dagger F - I| =", np.max(np.abs(f_dagger_f - identity2)))

eigvals_ff, eigvecs_ff = np.linalg.eigh(f_dagger_f)
off_one_index = int(np.argmax(np.abs(eigvals_ff - 1.0)))
probe_vector = eigvecs_ff[:, off_one_index]
sigma_probe = np.outer(probe_vector, probe_vector.conj())

print("Tr(sigma_probe)                  =", np.trace(sigma_probe))
print("Tr(F_AC sigma_probe F_AC^dagger) =", np.trace(f_ac @ sigma_probe @ f_ac.conj().T))
print()
print("FINITE_TRANSPORTER_IS_CHANNEL = NO")
print("NONUNITARY_PAIRWISE_TRANSPORT != CPTP_DYNAMICS")
````

## 19. Covariance locale sous `U_B`

`U_B` déterministe non trivial (`NON_NORMATIVE_QUALIFICATION_FIXTURE`),
