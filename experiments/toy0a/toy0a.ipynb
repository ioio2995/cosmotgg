## 1. Question de cette expérience

`toy0a` compare trois situations sur le même système biparti
`H_A ⊗ H_B = C^2 ⊗ C^2`, à marginales `ρ_A`, `ρ_B` **contrôlées** (fixées
identiques dans les trois cas) :

- **N0** : état produit (`PRODUCT_TRIVIAL_REGIME`) ;
- **N1** : état corrélé **commutant** (`COMMUTING_CORRELATED_REGIME`) ;
- **N2** : état corrélé **non commutant** (`NONCOMMUTING_CORRELATED_REGIME`).

Question de qualification posée par cette expérience :

> Le cocycle fini de Connes distingue-t-il une structure relationnelle
> plus fine que le simple fait « être corrélé » ou « ne pas être
> corrélé » ?

Cette question n'est **pas** formulée comme un test T1
(`docs/model/hypothesis.md`). `toy0a` est une qualification, pas une
opérationnalisation de T1 (`docs/toy-models/toy0a/specification.md` §1).

## 2. Famille d'états

La famille d'états canonique de `model0a`
(`TWO_QUBIT_FIXED_MARGINAL_CORRELATION_FAMILY`,
`docs/toy-models/toy0a/specification.md` §3) est paramétrée par
`(a, b, c, η)` :

- `a`, `b` fixent les marginales : `ρ_A = diag(a, 1-a)`, `ρ_B = diag(b, 1-b)` ;
- `c` module la corrélation **diagonale** (commutante) sans changer les marginales ;
- `η` ajoute une corrélation **hors-diagonale** (potentiellement non commutante).

Point important : à `a, b` fixés, faire varier `c` et `η` **ne change
pas** les marginales `ρ_A`, `ρ_B`. Cela permettra d'isoler l'effet de la
structure de corrélation, indépendamment d'un changement des marges.

````python
# Import du code testé (aucune fonction scientifique locale).
from cosmotgg.models.model0a.states import two_qubit_fixed_marginal_correlation_state

# NON_NORMATIVE_QUALIFICATION_FIXTURE : valeurs d'exemple de qualification
# uniquement. Elles ne constituent PAS STATE_PARAMETER_VALUES au sens
# scientifique de docs/toy-models/toy0a/specification.md (§16, OPEN).
STATE_PARAMETER_VALUES = "OPEN"
MODULAR_PARAMETER_DOMAIN = "OPEN"
NUMERICAL_TOLERANCES = "OPEN"

print(f"STATE_PARAMETER_VALUES   = {STATE_PARAMETER_VALUES}")
print(f"MODULAR_PARAMETER_DOMAIN = {MODULAR_PARAMETER_DOMAIN}")
print(f"NUMERICAL_TOLERANCES     = {NUMERICAL_TOLERANCES}")

# NON_NORMATIVE_QUALIFICATION_FIXTURE
a, b = 0.3, 0.4
fixture_params = {
    "N0": {"c": 0.0, "eta": 0.0},
    "N1": {"c": 0.02, "eta": 0.0},
    "N2": {"c": 0.02, "eta": 0.05},
}

# NON_NORMATIVE_QUALIFICATION_NUMERICS : tolérance d'exécution utilisée
# pour appeler les primitives core ci-dessous. Ce n'est ni une tolérance
# normative du modèle, ni un seuil scientifique, ni un critère
# d'acceptation, ni un paramètre confirmatoire.
QUALIFICATION_TOLERANCE = 1e-9
core_tolerances = dict(
    hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE,
    positivity_tolerance=QUALIFICATION_TOLERANCE,
)

print()
print(f"a = {a}, b = {b}  (NON_NORMATIVE_QUALIFICATION_FIXTURE)")
print(f"QUALIFICATION_TOLERANCE = {QUALIFICATION_TOLERANCE}  (NON_NORMATIVE_QUALIFICATION_NUMERICS)")
````

## 3. Construction N0 / N1 / N2

On construit maintenant les trois états `ρ_AB` avec les fixtures de
qualification ci-dessus, en appelant directement
`two_qubit_fixed_marginal_correlation_state` (aucune construction
manuelle de matrice dans ce notebook).

````python
# Helper purement présentationnel : arrondi d'affichage d'une matrice.
# Ne transforme aucune donnée scientifique ; sert uniquement à la lecture.
def display_matrix(label, matrix, decimals=4):
    print(f"{label} =")
    print(np.round(matrix, decimals))
    print()


rho = {
    name: two_qubit_fixed_marginal_correlation_state(a, b, p["c"], p["eta"])
    for name, p in fixture_params.items()
}

for name in ("N0", "N1", "N2"):
    display_matrix(f"rho_{name}", rho[name])
````

## 4. Marginales fixes

On extrait les marginales `ρ_A`, `ρ_B` de chaque état via
`cosmotgg.core.states.partial_trace`. Ce contrôle est important : il
garantit que toute différence observée plus loin (information mutuelle,
`R_AB`, cocycle, `C_AB`, `G`) provient de la structure de corrélation
(`c`, `η`), **pas** d'un changement simultané des marginales.

````python
from cosmotgg.core.states import partial_trace

rho_a = {name: partial_trace(rho[name], dimensions=(2, 2), keep=[0]) for name in rho}
rho_b = {name: partial_trace(rho[name], dimensions=(2, 2), keep=[1]) for name in rho}

for name in ("N0", "N1", "N2"):
    display_matrix(f"rho_A [{name}]", rho_a[name])
    display_matrix(f"rho_B [{name}]", rho_b[name])

marginals_a_identical = all(
    np.allclose(rho_a["N0"], rho_a[name]) for name in ("N1", "N2")
)
marginals_b_identical = all(
    np.allclose(rho_b["N0"], rho_b[name]) for name in ("N1", "N2")
)
print(f"rho_A identique dans N0/N1/N2 : {marginals_a_identical}")
print(f"rho_B identique dans N0/N1/N2 : {marginals_b_identical}")
````

## 5. État de référence `σ_AB`

L'état de référence `σ_AB = ρ_A ⊗ ρ_B` (`docs/toy-models/toy0a/specification.md`
§4) est obtenu via `model0a_reference_state` (assemblage `core`, aucune
logique nouvelle). À `a, b` fixés, il doit être identique dans les trois
régimes, exactement parce que les marginales le sont (§4 ci-dessus).

````python
from cosmotgg.models.model0a.diagnostics import model0a_reference_state

sigma = {name: model0a_reference_state(rho[name]) for name in rho}

for name in ("N0", "N1", "N2"):
    display_matrix(f"sigma_AB [{name}]", sigma[name])

sigma_identical = all(np.allclose(sigma["N0"], sigma[name]) for name in ("N1", "N2"))
print(f"sigma_AB identique dans N0/N1/N2 : {sigma_identical}")
````

## 6. Corrélations : information mutuelle

`I(A:B) = S(ρ_A) + S(ρ_B) - S(ρ_AB)` via
`cosmotgg.core.information.mutual_information`.

Interprétation autorisée : `N0` ne porte pas de corrélation ; `N1` et
`N2` en portent. **Point narratif important** : l'information mutuelle,
seule, ne sépare pas une corrélation *commutante* d'une corrélation
*non commutante* — c'est précisément ce que les sections suivantes
(`R_AB`, cocycle, `C_AB`, `G`) vont chercher à distinguer.

````python
from cosmotgg.core.information import mutual_information

mutual_info = {
    name: mutual_information(rho[name], dimensions=(2, 2), **core_tolerances)
    for name in rho
}

for name in ("N0", "N1", "N2"):
    print(f"I(A:B) [{name}] = {mutual_info[name]}")
````

## 7. Premier objet relationnel `R_AB`

$$
\mathcal R_{AB} = \ln(\rho_{AB}) - \ln(\rho_A \otimes \rho_B)
$$

via `cosmotgg.core.information.log_density_difference(rho_AB, sigma_AB, ...)`
(`docs/model/hypothesis.md` §5, `docs/toy-models/toy0a/specification.md`
§6). C'est le premier ordre analytique du cocycle :

$$
\left.\frac{d}{ds} v_s\right|_{s=0} = -i\,\mathcal R_{AB}.
$$

`R_AB` n'est appelé nulle part « générateur de temps » (`time
generator`) : c'est un opérateur relationnel de premier ordre, pas une
identification physique.

````python
from cosmotgg.core.information import log_density_difference

r_ab = {
    name: log_density_difference(rho[name], sigma[name], **core_tolerances)
    for name in rho
}

for name in ("N0", "N1", "N2"):
    display_matrix(f"R_AB [{name}]", r_ab[name])
    print(f"max|R_AB| [{name}] = {np.max(np.abs(r_ab[name]))}")
    print()
````

## 8. Cocycle fini `v_s`

$$
v_s(\rho,\sigma) = \rho^{-is}\,\sigma^{+is}
$$

via `cosmotgg.core.modular.finite_connes_cocycle(rho_AB, sigma_AB, s, ...)`,
pour une valeur de qualification déterministe `s`
(`NON_NORMATIVE_QUALIFICATION_FIXTURE`, aucun domaine officiel de `s`
n'est défini par ce notebook — `MODULAR_PARAMETER_DOMAIN` reste `OPEN`).

Attendu qualitatif :

- `N0` → `v_s` proche de l'identité ;
- `N1` → structure non triviale mais issue d'un régime commutant ;
- `N2` → structure non triviale issue d'un régime non commutant.

Une seule valeur de `s` ne doit **pas** être transformée en conclusion
générale sur tout le domaine modulaire.

````python
from cosmotgg.core.modular import finite_connes_cocycle

# NON_NORMATIVE_QUALIFICATION_FIXTURE
s = 0.4

v_s = {name: finite_connes_cocycle(rho[name], sigma[name], s, **core_tolerances) for name in rho}

identity_4 = np.eye(4)
for name in ("N0", "N1", "N2"):
    display_matrix(f"v_s [{name}] (s={s})", v_s[name])
    print(f"max|v_s - I| [{name}] = {np.max(np.abs(v_s[name] - identity_4))}")
    print()
````

## 9. Au-delà du premier ordre : `C_AB`

$$
C_{AB} = [\ln(\rho_{AB}),\ \ln(\sigma_{AB})]
$$

via `cosmotgg.models.model0a.diagnostics.log_commutator_obstruction`
(`docs/toy-models/toy0a/specification.md` §9.4). Rappel de l'équivalence
enregistrée pour des états fidèles :

$$
C_{AB} = 0 \iff [\rho_{AB},\sigma_{AB}] = 0.
$$

C'est la distinction essentielle que ce notebook cherche à faire
apparaître : `N1` est corrélé **mais** `C_AB = 0` ; `N2` est corrélé
**et** `C_AB ≠ 0`. `C_AB` n'est ni une courbure, ni un générateur
temporel, ni un générateur causal (§9.4 de la spécification).

````python
from cosmotgg.models.model0a.diagnostics import log_commutator_obstruction

c_ab = {name: log_commutator_obstruction(rho[name], **core_tolerances) for name in rho}

for name in ("N0", "N1", "N2"):
    display_matrix(f"C_AB [{name}]", c_ab[name])
    print(f"max|C_AB| [{name}] = {np.max(np.abs(c_ab[name]))}")
    print()
````

## 10. Défaut de groupe ordinaire `G(s1, s2)`

$$
G(s_1,s_2) = v_{s_1+s_2} - v_{s_1}v_{s_2}
$$

via `cosmotgg.models.model0a.diagnostics.ordinary_group_defect`
(`docs/toy-models/toy0a/specification.md` §9.5), pour un couple de
qualification déterministe `(s1, s2)`
(`NON_NORMATIVE_QUALIFICATION_FIXTURE`).

**Important :** le cocycle de Connes satisfait sa propre identité de
cocycle (`docs/toy-models/toy0a/specification.md` §7) ; ceci n'est
**pas** un défaut de cette identité. `G` mesure uniquement si le cocycle
satisfait, en plus, la propriété — distincte et plus forte — d'un groupe
unitaire ordinaire sous multiplication directe.

Le commentaire ci-dessous porte uniquement sur les valeurs réellement
calculées pour cette fixture précise.

````python
from cosmotgg.models.model0a.diagnostics import ordinary_group_defect

# NON_NORMATIVE_QUALIFICATION_FIXTURE
s1, s2 = 0.3, 0.4

g_s1_s2 = {name: ordinary_group_defect(rho[name], s1, s2, **core_tolerances) for name in rho}

for name in ("N0", "N1", "N2"):
    display_matrix(f"G(s1,s2) [{name}] (s1={s1}, s2={s2})", g_s1_s2[name])
    print(f"max|G| [{name}] = {np.max(np.abs(g_s1_s2[name]))}")
    print()
````

## 11. Relation entre `C_AB` et `G`

Identité analytique déjà enregistrée
(`docs/toy-models/toy0a/specification.md` §9.6) :

$$
\left.\frac{\partial^2 G}{\partial s_1\,\partial s_2}\right|_{(0,0)}
=
[\ln(\rho_{AB}),\ln(\sigma_{AB})]
=
C_{AB}.
$$

Cette identité relie l'obstruction statique `C_AB` (§9 ci-dessus) à la
première obstruction locale à la propriété de groupe ordinaire du
cocycle (§10 ci-dessus). Elle possède déjà un test de régression
numérique automatisé et indépendant dans
`tests/models/model0a/test_diagnostics.py`
(`test_ordinary_group_defect_local_identity_matches_log_commutator_regression`).
Ce notebook ne duplique pas cette logique de différence finie.
