## 1. Question de cette expérience

`toy0a` compare trois situations sur le même système biparti
`H_A ⊗ H_B = C^2 ⊗ C^2`, à marginales `ρ_A`, `ρ_B` **contrôlées** (fixées
identiques dans les trois cas) :

- **N0** : état produit (`PRODUCT_TRIVIAL_REGIME`) ;
- **N1** : état corrélé **commutant** (`COMMUTING_CORRELATED_REGIME`) ;
- **N2** : état corrélé **non commutant** (`NONCOMMUTING_CORRELATED_REGIME`).

Question de qualification posée par cette expérience :

> Le cocycle fini de Connes distingue-t-il une structure relationnelle
> plus fine que le simple fait « être corrélé » ou « ne pas être
> corrélé » ?

Cette question n'est **pas** formulée comme un test T1
(`docs/model/hypothesis.md`). `toy0a` est une qualification, pas une
opérationnalisation de T1 (`docs/toy-models/toy0a/specification.md` §1).

## 2. Famille d'états

La famille d'états canonique de `model0a`
(`TWO_QUBIT_FIXED_MARGINAL_CORRELATION_FAMILY`,
`docs/toy-models/toy0a/specification.md` §3) est paramétrée par
`(a, b, c, η)` :

- `a`, `b` fixent les marginales : `ρ_A = diag(a, 1-a)`, `ρ_B = diag(b, 1-b)` ;
- `c` module la corrélation **diagonale** (commutante) sans changer les marginales ;
- `η` ajoute une corrélation **hors-diagonale** (potentiellement non commutante).

Point important : à `a, b` fixés, faire varier `c` et `η` **ne change
pas** les marginales `ρ_A`, `ρ_B`. Cela permettra d'isoler l'effet de la
structure de corrélation, indépendamment d'un changement des marges.

````python
# Import du code testé (aucune fonction scientifique locale).
from cosmotgg.models.model0a.states import two_qubit_fixed_marginal_correlation_state

# NON_NORMATIVE_QUALIFICATION_FIXTURE : valeurs d'exemple de qualification
# uniquement. Elles ne constituent PAS STATE_PARAMETER_VALUES au sens
# scientifique de docs/toy-models/toy0a/specification.md (§16, OPEN).
STATE_PARAMETER_VALUES = "OPEN"
MODULAR_PARAMETER_DOMAIN = "OPEN"
NUMERICAL_TOLERANCES = "OPEN"

print(f"STATE_PARAMETER_VALUES   = {STATE_PARAMETER_VALUES}")
print(f"MODULAR_PARAMETER_DOMAIN = {MODULAR_PARAMETER_DOMAIN}")
print(f"NUMERICAL_TOLERANCES     = {NUMERICAL_TOLERANCES}")

# NON_NORMATIVE_QUALIFICATION_FIXTURE
a, b = 0.3, 0.4
fixture_params = {
    "N0": {"c": 0.0, "eta": 0.0},
    "N1": {"c": 0.02, "eta": 0.0},
    "N2": {"c": 0.02, "eta": 0.05},
}

# NON_NORMATIVE_QUALIFICATION_NUMERICS : tolérance d'exécution utilisée
# pour appeler les primitives core ci-dessous. Ce n'est ni une tolérance
# normative du modèle, ni un seuil scientifique, ni un critère
# d'acceptation, ni un paramètre confirmatoire.
QUALIFICATION_TOLERANCE = 1e-9
core_tolerances = dict(
    hermiticity_tolerance=QUALIFICATION_TOLERANCE,
    trace_tolerance=QUALIFICATION_TOLERANCE,
    positivity_tolerance=QUALIFICATION_TOLERANCE,
)

print()
print(f"a = {a}, b = {b}  (NON_NORMATIVE_QUALIFICATION_FIXTURE)")
print(f"QUALIFICATION_TOLERANCE = {QUALIFICATION_TOLERANCE}  (NON_NORMATIVE_QUALIFICATION_NUMERICS)")
````

## 3. Construction N0 / N1 / N2

On construit maintenant les trois états `ρ_AB` avec les fixtures de
qualification ci-dessus, en appelant directement
`two_qubit_fixed_marginal_correlation_state` (aucune construction
manuelle de matrice dans ce notebook).

````python
# Helper purement présentationnel : arrondi d'affichage d'une matrice.
# Ne transforme aucune donnée scientifique ; sert uniquement à la lecture.
def display_matrix(label, matrix, decimals=4):
    print(f"{label} =")
    print(np.round(matrix, decimals))
    print()


rho = {
    name: two_qubit_fixed_marginal_correlation_state(a, b, p["c"], p["eta"])
    for name, p in fixture_params.items()
}

for name in ("N0", "N1", "N2"):
    display_matrix(f"rho_{name}", rho[name])
````

## 4. Marginales fixes

On extrait les marginales `ρ_A`, `ρ_B` de chaque état via
`cosmotgg.core.states.partial_trace`. Ce contrôle est important : il
garantit que toute différence observée plus loin (information mutuelle,
`R_AB`, cocycle, `C_AB`, `G`) provient de la structure de corrélation
(`c`, `η`), **pas** d'un changement simultané des marginales.

````python
from cosmotgg.core.states import partial_trace

rho_a = {name: partial_trace(rho[name], dimensions=(2, 2), keep=[0]) for name in rho}
rho_b = {name: partial_trace(rho[name], dimensions=(2, 2), keep=[1]) for name in rho}

for name in ("N0", "N1", "N2"):
    display_matrix(f"rho_A [{name}]", rho_a[name])
    display_matrix(f"rho_B [{name}]", rho_b[name])

marginals_a_identical = all(
    np.allclose(rho_a["N0"], rho_a[name]) for name in ("N1", "N2")
)
marginals_b_identical = all(
    np.allclose(rho_b["N0"], rho_b[name]) for name in ("N1", "N2")
)
print(f"rho_A identique dans N0/N1/N2 : {marginals_a_identical}")
print(f"rho_B identique dans N0/N1/N2 : {marginals_b_identical}")
````

## 5. État de référence `σ_AB`

L'état de référence `σ_AB = ρ_A ⊗ ρ_B` (`docs/toy-models/toy0a/specification.md`
§4) est obtenu via `model0a_reference_state` (assemblage `core`, aucune
logique nouvelle). À `a, b` fixés, il doit être identique dans les trois
régimes, exactement parce que les marginales le sont (§4 ci-dessus).

````python
from cosmotgg.models.model0a.diagnostics import model0a_reference_state

sigma = {name: model0a_reference_state(rho[name]) for name in rho}

for name in ("N0", "N1", "N2"):
    display_matrix(f"sigma_AB [{name}]", sigma[name])

sigma_identical = all(np.allclose(sigma["N0"], sigma[name]) for name in ("N1", "N2"))
print(f"sigma_AB identique dans N0/N1/N2 : {sigma_identical}")
````

## 6. Corrélations : information mutuelle

`I(A:B) = S(ρ_A) + S(ρ_B) - S(ρ_AB)` via
`cosmotgg.core.information.mutual_information`.

Interprétation autorisée : `N0` ne porte pas de corrélation ; `N1` et
`N2` en portent. **Point narratif important** : l'information mutuelle,
seule, ne sépare pas une corrélation *commutante* d'une corrélation
*non commutante* — c'est précisément ce que les sections suivantes
(`R_AB`, cocycle, `C_AB`, `G`) vont chercher à distinguer.

````python
from cosmotgg.core.information import mutual_information

mutual_info = {
    name: mutual_information(rho[name], dimensions=(2, 2), **core_tolerances)
    for name in rho
}

for name in ("N0", "N1", "N2"):
    print(f"I(A:B) [{name}] = {mutual_info[name]}")
````

## 7. Premier objet relationnel `R_AB`

$$
\mathcal R_{AB} = \ln(\rho_{AB}) - \ln(\rho_A \otimes \rho_B)
$$

via `cosmotgg.core.information.log_density_difference(rho_AB, sigma_AB, ...)`
(`docs/model/hypothesis.md` §5, `docs/toy-models/toy0a/specification.md`
§6). C'est le premier ordre analytique du cocycle :

$$
\left.\frac{d}{ds} v_s\right|_{s=0} = -i\,\mathcal R_{AB}.
$$

`R_AB` n'est appelé nulle part « générateur de temps » (`time
generator`) : c'est un opérateur relationnel de premier ordre, pas une
identification physique.

````python
from cosmotgg.core.information import log_density_difference

r_ab = {
    name: log_density_difference(rho[name], sigma[name], **core_tolerances)
    for name in rho
}

for name in ("N0", "N1", "N2"):
    display_matrix(f"R_AB [{name}]", r_ab[name])
    print(f"max|R_AB| [{name}] = {np.max(np.abs(r_ab[name]))}")
    print()
````

## 8. Cocycle fini `v_s`

$$
v_s(\rho,\sigma) = \rho^{-is}\,\sigma^{+is}
$$

via `cosmotgg.core.modular.finite_connes_cocycle(rho_AB, sigma_AB, s, ...)`,
pour une valeur de qualification déterministe `s`
(`NON_NORMATIVE_QUALIFICATION_FIXTURE`, aucun domaine officiel de `s`
n'est défini par ce notebook — `MODULAR_PARAMETER_DOMAIN` reste `OPEN`).

Attendu qualitatif :

- `N0` → `v_s` proche de l'identité ;
- `N1` → structure non triviale mais issue d'un régime commutant ;
- `N2` → structure non triviale issue d'un régime non commutant.

Une seule valeur de `s` ne doit **pas** être transformée en conclusion
générale sur tout le domaine modulaire.

````python
from cosmotgg.core.modular import finite_connes_cocycle

# NON_NORMATIVE_QUALIFICATION_FIXTURE
s = 0.4

v_s = {name: finite_connes_cocycle(rho[name], sigma[name], s, **core_tolerances) for name in rho}

identity_4 = np.eye(4)
for name in ("N0", "N1", "N2"):
    display_matrix(f"v_s [{name}] (s={s})", v_s[name])
    print(f"max|v_s - I| [{name}] = {np.max(np.abs(v_s[name] - identity_4))}")
    print()
````

## 9. Au-delà du premier ordre : `C_AB`

$$
C_{AB} = [\ln(\rho_{AB}),\ \ln(\sigma_{AB})]
$$

via `cosmotgg.models.model0a.diagnostics.log_commutator_obstruction`
(`docs/toy-models/toy0a/specification.md` §9.4). Rappel de l'équivalence
enregistrée pour des états fidèles :

$$
C_{AB} = 0 \iff [\rho_{AB},\sigma_{AB}] = 0.
$$

C'est la distinction essentielle que ce notebook cherche à faire
apparaître : `N1` est corrélé **mais** `C_AB = 0` ; `N2` est corrélé
**et** `C_AB ≠ 0`. `C_AB` n'est ni une courbure, ni un générateur
temporel, ni un générateur causal (§9.4 de la spécification).

````python
from cosmotgg.models.model0a.diagnostics import log_commutator_obstruction

c_ab = {name: log_commutator_obstruction(rho[name], **core_tolerances) for name in rho}

for name in ("N0", "N1", "N2"):
    display_matrix(f"C_AB [{name}]", c_ab[name])
    print(f"max|C_AB| [{name}] = {np.max(np.abs(c_ab[name]))}")
    print()
````

## 15. Covariance sous changement de bases locales

Pourquoi ce contrôle est nécessaire : les distinctions N0/N1/N2 et les
diagnostics modulaires (`R_AB`, `v_s`, `C_AB`, `G`) observés plus haut ne
doivent pas être un artefact du choix particulier de la base
`|00⟩,|01⟩,|10⟩,|11⟩`. On vérifie donc leur comportement sous un
changement de bases **locales** :

$$
U = U_A \otimes U_B,
\qquad
\rho' = U\,\rho\,U^\dagger.
$$

Un tel `U` préserve exactement la décomposition tensorielle `A|B` : il ne
fait que changer la base à l'intérieur de chaque facteur `H_A`, `H_B`
séparément.

```text
LOCAL_BASIS_COVARIANCE != ARBITRARY_GLOBAL_UNITARY_COVARIANCE
```

**Frontière explicite :** un unitaire global général (entanglant) peut
changer la structure même des sous-systèmes par rapport à laquelle
`ρ_A`, `ρ_B`, `σ_AB` sont définis. Ce contrôle **ne teste pas**
l'invariance sous un changement arbitraire de factorisation tensorielle
— seulement sous `U_A ⊗ U_B`.

````python
# NON_NORMATIVE_QUALIFICATION_FIXTURE : unitaires locaux déterministes,
# sans statut physique privilégié, choisis pour sortir de la base
# canonique (non diagonaux).
sqrt2 = np.sqrt(2.0)
U_A = (1.0 / sqrt2) * np.array([[1.0, 1.0], [1.0, -1.0]], dtype=complex)
U_B = (1.0 / sqrt2) * np.array([[1.0, 1.0j], [1.0j, 1.0]], dtype=complex)

print("max|U_A^dagger U_A - I| =", np.max(np.abs(U_A.conj().T @ U_A - np.eye(2))))
print("max|U_B^dagger U_B - I| =", np.max(np.abs(U_B.conj().T @ U_B - np.eye(2))))

U_local = np.kron(U_A, U_B)
print("max|U^dagger U - I|     =", np.max(np.abs(U_local.conj().T @ U_local - np.eye(4))))
````

On applique `ρ' = U ρ U†` aux trois régimes `N0`/`N1`/`N2` déjà
construits (§3), puis on recalcule chaque quantité **depuis `ρ'`** avec
le même code testé qu'aux sections précédentes (aucune nouvelle
fonction), et on compare au transporté direct `U · (quantité) · U†`.

````python
rho_transformed = {
    name: U_local @ rho[name] @ U_local.conj().T for name in rho
}

residuals = {}
for name in ("N0", "N1", "N2"):
    sigma_t = model0a_reference_state(rho_transformed[name])
    r_ab_t = log_density_difference(rho_transformed[name], sigma_t, **core_tolerances)
    v_s_t = finite_connes_cocycle(rho_transformed[name], sigma_t, s, **core_tolerances)
    c_ab_t = log_commutator_obstruction(rho_transformed[name], **core_tolerances)
    g_t = ordinary_group_defect(rho_transformed[name], s1, s2, **core_tolerances)
    i_t = mutual_information(rho_transformed[name], dimensions=(2, 2), **core_tolerances)

    # U_local @ (...) @ U_local.conj().T below is an explicit configuration
    # step of this covariance control (transporting the untransformed
    # quantity by the same fixed local unitary), not a reusable scientific
    # primitive of the model.
    residuals[name] = {
        "sigma": np.max(np.abs(sigma_t - U_local @ sigma[name] @ U_local.conj().T)),
        "R_AB": np.max(np.abs(r_ab_t - U_local @ r_ab[name] @ U_local.conj().T)),
        "v_s": np.max(np.abs(v_s_t - U_local @ v_s[name] @ U_local.conj().T)),
        "C_AB": np.max(np.abs(c_ab_t - U_local @ c_ab[name] @ U_local.conj().T)),
        "G": np.max(np.abs(g_t - U_local @ g_s1_s2[name] @ U_local.conj().T)),
        "I": abs(i_t - mutual_info[name]),
    }

print(f"{'régime':<6}{'max|sigma diff|':>18}{'max|R diff|':>14}{'max|v_s diff|':>16}{'max|C diff|':>14}{'max|G diff|':>14}{'I diff':>12}")
for name in ("N0", "N1", "N2"):
    r = residuals[name]
    print(
        f"{name:<6}"
        f"{r['sigma']:>18.3g}"
        f"{r['R_AB']:>14.3g}"
        f"{r['v_s']:>16.3g}"
        f"{r['C_AB']:>14.3g}"
        f"{r['G']:>14.3g}"
        f"{r['I']:>12.3g}"
    )
````

Ces résidus (`NON_NORMATIVE_QUALIFICATION_NUMERICS`, purement
diagnostiques d'exécution) ne constituent aucun nouveau scalaire
scientifique : ce sont des différences numériques entre deux calculs
issus des mêmes primitives testées, illustrant les identités de
covariance vérifiées dans ce contrôle (`σ' = UσU†`, `I'=I`,
`R'=URU†`, `v'_s=Uv_sU†`, `C'=UCU†`, `G'=UGU†`).

On vérifie enfin que la classification structurelle N0/N1/N2 est
préservée après transformation.

````python
for name in ("N0", "N1", "N2"):
    sigma_t = model0a_reference_state(rho_transformed[name])
    commutator_t = rho_transformed[name] @ sigma_t - sigma_t @ rho_transformed[name]
    print(f"{name}: rho' ~= sigma' -> {np.allclose(rho_transformed[name], sigma_t)}"
          f" ; max|[rho',sigma']| = {np.max(np.abs(commutator_t)):.6g}")
````

**Interprétation maximale autorisée :** si ces contrôles passent, la
distinction N0/N1/N2 et les diagnostics modulaires associés (`R_AB`,
`v_s`, `C_AB`, `G`) ne dépendent pas du choix de bases locales dans
`H_A` et `H_B` — uniquement de `LOCAL_PRODUCT_UNITARY_COVARIANCE`. Ceci
ne permet de conclure ni une `physical covariance`, ni une
`spacetime covariance`, ni une `Lorentz covariance`, ni une
`gauge invariance`, ni une `general covariance`.

La structure qualifiée n'est donc pas un artefact de base locale ; mais
cela ne ferme aucun critère scientifique :

```text
T1_NONTRIVIALITY_CRITERION = OPEN
```

La question ouverte de la §14 (construction d'un changement **RELATIF**
entre sous-structures, sans réduire `s` à une horloge physique par
simple élimination paramétrique) reste entièrement posée et n'est pas
close par ce contrôle de covariance.

## 16. Contrôles négatifs

Cette section teste explicitement trois lectures fausses possibles des
résultats précédents, pour éviter qu'ils ne soient sur-interprétés.

### 16.1 Une cohérence hors diagonale ne suffit pas

`η ≠ 0` seul **ne suffit pas** à produire `[ρ,σ] ≠ 0`, `C_AB ≠ 0` ou
`G ≠ 0`.

Fixture (`NON_NORMATIVE_NEGATIVE_CONTROL_FIXTURE`) : `a=0.3`, `b=0.7`
(donc `a+b=1`), `c=0`, `η=0.05`.

````python
# NON_NORMATIVE_NEGATIVE_CONTROL_FIXTURE
a_nc1, b_nc1, c_nc1, eta_nc1 = 0.3, 0.7, 0.0, 0.05
print(f"a + b = {a_nc1 + b_nc1}")

rho_nc1 = two_qubit_fixed_marginal_correlation_state(a_nc1, b_nc1, c_nc1, eta_nc1)
sigma_nc1 = model0a_reference_state(rho_nc1)

display_matrix("rho_nc1", rho_nc1)
display_matrix("sigma_nc1", sigma_nc1)

i_nc1 = mutual_information(rho_nc1, dimensions=(2, 2), **core_tolerances)
commutator_nc1 = rho_nc1 @ sigma_nc1 - sigma_nc1 @ rho_nc1
c_ab_nc1 = log_commutator_obstruction(rho_nc1, **core_tolerances)
g_nc1 = ordinary_group_defect(rho_nc1, s1, s2, **core_tolerances)

print(f"I(A:B)            = {i_nc1}")
print(f"max|[rho,sigma]|  = {np.max(np.abs(commutator_nc1))}")
print(f"max|C_AB|         = {np.max(np.abs(c_ab_nc1))}")
print(f"max|G(s1,s2)|     = {np.max(np.abs(g_nc1))}   (s1={s1}, s2={s2})")
````

On observe `η ≠ 0`, `ρ ≠ σ`, `I(A:B) > 0`, mais un commutateur, un
`C_AB` et un `G` numériquement nuls.

Explication analytique : lorsque `a+b=1`, `σ_00 = a·b = (1-a)(1-b) = σ_11` :
la référence `σ` a une **dégénérescence** entre les blocs `|00⟩` et
`|11⟩`. La cohérence `η` du bloc `ρ` agit alors *à l'intérieur* de ce
sous-espace dégénéré de `σ`, ce qui suffit à préserver la commutation
`[ρ,σ]=0`, indépendamment de la valeur de `η`.

**Conclusion :** `off-diagonal coherence != noncommutativity`. Cet état
appartient structurellement au régime `COMMUTING_CORRELATED_REGIME`
(N1) bien que `η ≠ 0` : la tranche canonique `η=0` (spécification §3.3)
est une construction **pratique** pour N1, pas une définition
**nécessaire** de N1.

### 16.2 La quantité de corrélation n'est pas le classifieur

On compare deux états à `a=0.3`, `b=0.4`
(`NON_NORMATIVE_NEGATIVE_CONTROL_FIXTURE`) :

- **commuting** : `c=0.08`, `η=0` ;
- **noncommuting** : `c=0.02`, `η=0.01`.

Aucun seuil sur `I(A:B)` n'est introduit ; les valeurs ci-dessous sont
calculées par le code, pas préenregistrées comme oracle.

````python
# NON_NORMATIVE_NEGATIVE_CONTROL_FIXTURE
a_nc2, b_nc2 = 0.3, 0.4

rho_nc2_commuting = two_qubit_fixed_marginal_correlation_state(a_nc2, b_nc2, 0.08, 0.0)
rho_nc2_noncommuting = two_qubit_fixed_marginal_correlation_state(a_nc2, b_nc2, 0.02, 0.01)

sigma_nc2_commuting = model0a_reference_state(rho_nc2_commuting)
sigma_nc2_noncommuting = model0a_reference_state(rho_nc2_noncommuting)

i_nc2_commuting = mutual_information(rho_nc2_commuting, dimensions=(2, 2), **core_tolerances)
i_nc2_noncommuting = mutual_information(rho_nc2_noncommuting, dimensions=(2, 2), **core_tolerances)

c_ab_nc2_commuting = log_commutator_obstruction(rho_nc2_commuting, **core_tolerances)
c_ab_nc2_noncommuting = log_commutator_obstruction(rho_nc2_noncommuting, **core_tolerances)

g_nc2_commuting = ordinary_group_defect(rho_nc2_commuting, s1, s2, **core_tolerances)
g_nc2_noncommuting = ordinary_group_defect(rho_nc2_noncommuting, s1, s2, **core_tolerances)

print(f"{'fixture':<14}{'I(A:B)':>12}{'max|C_AB|':>14}{'max|G|':>14}")
print(
    f"{'commuting':<14}"
    f"{i_nc2_commuting:>12.6g}"
    f"{np.max(np.abs(c_ab_nc2_commuting)):>14.6g}"
    f"{np.max(np.abs(g_nc2_commuting)):>14.6g}"
)
print(
    f"{'noncommuting':<14}"
    f"{i_nc2_noncommuting:>12.6g}"
    f"{np.max(np.abs(c_ab_nc2_noncommuting)):>14.6g}"
    f"{np.max(np.abs(g_nc2_noncommuting)):>14.6g}"
)
print()
print(f"I_commuting > I_noncommuting : {i_nc2_commuting > i_nc2_noncommuting}")
````

L'état **commutant** porte une information mutuelle strictement **plus
grande** que l'état non commutant, tout en ayant `C_AB = 0` et `G = 0` là
où l'état non commutant a `C_AB ≠ 0` et `G ≠ 0` (pour ce couple).

**Conclusion :** `I(A:B)` détecte la corrélation, pas la classe
commutante/non commutante. Aucun seuil, aucune normalisation, aucun
score composite n'est introduit sur `I(A:B)`.

### 16.3 `G` peut s'annuler ponctuellement en N2

Sur la fixture N2 déjà construite (§3), `C_AB ≠ 0` (régime non
commutant). Pourtant `v_0 = I` (§8) implique exactement :

$$
G(0,s_2) = v_{s_2} - v_0 v_{s_2} = 0,
\qquad
G(s_1,0) = v_{s_1} - v_{s_1} v_0 = 0
$$

pour tout `s_1`, `s_2` — des zéros **nécessaires** sur les axes, pas des
coïncidences numériques.

````python
c_ab_n2 = c_ab["N2"]  # déjà calculé en section 9
print(f"max|C_AB| [N2] = {np.max(np.abs(c_ab_n2))}")

g_zero_s1 = ordinary_group_defect(rho["N2"], 0.0, s2, **core_tolerances)
g_zero_s2 = ordinary_group_defect(rho["N2"], s1, 0.0, **core_tolerances)
g_generic = g_s1_s2["N2"]  # déjà calculé en section 10, (s1, s2) = (0.3, 0.4)

print(f"max|G(0, s2)|  = {np.max(np.abs(g_zero_s1))}")
print(f"max|G(s1, 0)|  = {np.max(np.abs(g_zero_s2))}")
print(f"max|G(s1, s2)| = {np.max(np.abs(g_generic))}   (s1={s1}, s2={s2}, couple générique)")
````

**Conclusion :** en N2, `G` est `NOT_IDENTICALLY_ZERO` (§9.7 de la
spécification), mais des zéros particuliers existent nécessairement,
notamment sur les axes `s1=0` ou `s2=0`. Il ne faut **jamais** écrire
`G ≠ 0 pour tout s1,s2`.
